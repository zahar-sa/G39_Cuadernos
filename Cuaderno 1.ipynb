## Probabilidad Laplaciana (Clásica)

### Definición

La probabilidad laplaciana, también llamada probabilidad clásica o a priori, se define como el cociente entre el número de casos favorables y el número de casos posibles, siempre que todos los resultados sean igualmente probables.

$$P(A) = \frac{\text{Casos favorables}}{\text{Casos posibles}}$$

### Condiciones de aplicación

| Condición | Explicación |
|-----------|-------------|
| Espacio muestral finito | El número total de resultados posibles es limitado y conocido |
| Equiprobabilidad | Todos los resultados tienen la misma probabilidad de ocurrir |
| Experimento reproducible | Se puede repetir el experimento en las mismas condiciones |

### Analogía

> "Es como tener una bolsa con 10 canicas del mismo tamaño, 3 rojas y 7 azules. La probabilidad de sacar una canica roja sin mirar es simplemente 3/10. No necesitas hacer experimentos, solo contar."

### Ejemplos reales

1. **Juegos de azar:** La probabilidad de obtener un 6 al lanzar un dado justo es 1/6.
2. **Lotería:** La probabilidad de que salga un número específico en una rifa de 100 boletos es 1/100.
3. **Nacimientos:** Suponiendo igual probabilidad, la probabilidad de que un bebé sea niño es 1/2.
4. **Mazos de cartas:** La probabilidad de sacar un as de una baraja de 52 cartas es 4/52 = 1/13.
5. **Ruleta europea:** La probabilidad de acertar un número específico es 1/37.

### Conceptos fundamentales

**Experimento aleatorio:** Proceso que produce resultados impredecibles.

**Espacio muestral (Ω):** Conjunto de todos los resultados posibles.

**Evento (A):** Subconjunto del espacio muestral (resultados que nos interesan).

### Propiedades básicas

1. $0 ≤ P(A) ≤ 1$
2. $P(Ω) = 1$ (probabilidad del suceso seguro)
3. $P(∅) = 0$ (probabilidad del suceso imposible)
4. $P(A^c) = 1 - P(A)$ (probabilidad del complemento)

### Ejemplo resuelto

**Problema:** Se lanzan dos dados justos. ¿Cuál es la probabilidad de que la suma sea 7?

**Solución:**
- Casos posibles: $6 × 6 = 36$
- Casos favorables (suma 7): (1,6), (2,5), (3,4), (4,3), (5,2), (6,1) → 6 casos
- $P(suma = 7) = 6/36 = 1/6 ≈ 0.1667$

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets
from itertools import product


print("Kernel listo y configurado")
```

```python
# Probabilidad Laplaciana - Demostración interactiva


def probabilidad_clasica(ejemplo="Dados - Suma 7"):
    """
    Demostración interactiva de probabilidad laplaciana
    """
    if ejemplo == "Dados - Suma 7":
        # Espacio muestral: lanzar dos dados
        caras = [1, 2, 3, 4, 5, 6]
        espacio = list(product(caras, repeat=2))
        total_casos = len(espacio)
        
        # Evento: suma igual a 7
        evento = [r for r in espacio if sum(r) == 7]
        favorables = len(evento)
        probabilidad = favorables / total_casos
        
        # Visualización
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        
        # Tabla de resultados
        axes[0].axis('tight')
        axes[0].axis('off')
        
        tabla_data = [[f"{r[0]}+{r[1]}={sum(r)}" for r in espacio[i:i+6]] for i in range(0, 36, 6)]
        table = axes[0].table(cellText=tabla_data, loc='center', cellLoc='center')
        table.auto_set_font_size(False)
        table.set_fontsize(8)
        axes[0].set_title("Espacio muestral (36 resultados posibles)", fontsize=10)
        
        # Resaltar casos favorables
        for i, row in enumerate(tabla_data):
            for j, cell in enumerate(row):
                if cell.split('=')[1] == '7':
                    table[(i, j)].set_facecolor('lightgreen')
        
        # Gráfico de barras
        sums = [sum(r) for r in espacio]
        valores, conteos = np.unique(sums, return_counts=True)
        prob_teorica = conteos / total_casos
        
        axes[1].bar(valores, prob_teorica, color='skyblue', edgecolor='black')
        axes[1].set_xlabel("Suma de los dados")
        axes[1].set_ylabel("Probabilidad")
        axes[1].set_title(f"Distribución de probabilidades\nP(suma=7) = {probabilidad:.4f}")
        axes[1].grid(axis='y', alpha=0.3)
        
        info = f"""
        📊 Cálculo de probabilidad laplaciana
        
        • Espacio muestral (Ω): {total_casos} resultados posibles
        • Evento A: "La suma de los dados es 7"
        • Casos favorables a A: {favorables}
        • Casos posibles: {total_casos}
        
        P(A) = {favorables} / {total_casos} = {probabilidad:.4f}
        """
        
    elif ejemplo == "Monedas - Al menos una cara":
        # Espacio muestral: lanzar 3 monedas
        monedas = ['C', 'S']
        espacio = list(product(monedas, repeat=3))
        total_casos = len(espacio)
        
        # Evento: al menos una cara
        evento = [r for r in espacio if 'C' in r]
        favorables = len(evento)
        probabilidad = favorables / total_casos
        
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        
        # Lista de resultados
        resultados = [''.join(r) for r in espacio]
        axes[0].axis('tight')
        axes[0].axis('off')
        axes[0].text(0.5, 0.5, "Espacio muestral (8 resultados):\n" + ", ".join(resultados), 
                    ha='center', va='center', transform=axes[0].transAxes, fontsize=12)
        
        # Gráfico de barras
        num_caras = [r.count('C') for r in espacio]
        valores, conteos = np.unique(num_caras, return_counts=True)
        prob_teorica = conteos / total_casos
        
        axes[1].bar(valores, prob_teorica, color='lightcoral', edgecolor='black')
        axes[1].set_xlabel("Número de caras")
        axes[1].set_ylabel("Probabilidad")
        axes[1].set_title(f"Distribución de probabilidades\nP(al menos una cara) = {probabilidad:.4f}")
        axes[1].grid(axis='y', alpha=0.3)
        
        info = f"""
        📊 Cálculo de probabilidad laplaciana
        
        • Espacio muestral (Ω): {total_casos} resultados posibles
        • Evento A: "Obtener al menos una cara en 3 lanzamientos"
        • Casos favorables a A: {favorables}
        • Casos posibles: {total_casos}
        
        P(A) = {favorables} / {total_casos} = {probabilidad:.4f}
        """
        
    else:  # Baraja
        total_casos = 52
        favorables = 4
        probabilidad = favorables / total_casos
        
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        
        axes[0].axis('tight')
        axes[0].axis('off')
        axes[0].text(0.5, 0.5, "Baraja española de 52 cartas\n4 palos: ♠ ♥ ♦ ♣\n13 cartas por palo", 
                    ha='center', va='center', transform=axes[0].transAxes, fontsize=12)
        
        # Gráfico de probabilidad
        categorias = ['Obtener As', 'Cualquier otra']
        probs = [probabilidad, 1 - probabilidad]
        axes[1].bar(categorias, probs, color=['gold', 'lightgray'], edgecolor='black')
        axes[1].set_ylabel("Probabilidad")
        axes[1].set_title(f"P(As) = {probabilidad:.4f}")
        axes[1].grid(axis='y', alpha=0.3)
        
        info = f"""
        📊 Cálculo de probabilidad laplaciana
        
        • Espacio muestral (Ω): {total_casos} cartas
        • Evento A: "Sacar un As"
        • Casos favorables a A: {favorables} (As de ♠, ♥, ♦, ♣)
        • Casos posibles: {total_casos}
        
        P(A) = {favorables} / {total_casos} = {probabilidad:.4f}
        """
    
    plt.suptitle('Probabilidad Laplaciana (Clásica)', fontsize=14, fontweight='bold')
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()

interact(probabilidad_clasica,
         ejemplo=widgets.Dropdown(
             options=['Dados - Suma 7', 'Monedas - Al menos una cara', 'Baraja - Obtener un As'],
             value='Dados - Suma 7',
             description='Ejemplo:',
             style={'description_width': 'initial'}
         ))
```

## Probabilidad Frecuentista

### Definición

La probabilidad frecuentista se define como el límite de la frecuencia relativa de un evento cuando el número de experimentos tiende a infinito.

$$P(A) = \lim_{n \to \infty} \frac{n_A}{n}$$

Donde:
- $n_A$ = número de veces que ocurre el evento A
- $n$ = número total de experimentos realizados

### Relación con la Ley de los Grandes Números

> "Al repetir un experimento muchas veces, la frecuencia relativa de un evento se aproxima a su probabilidad teórica."

### Analogía

> "Si lanzas una moneda 10 veces y obtienes 6 caras, la frecuencia relativa es 0.6. Pero si la lanzas 10,000 veces, esperarías que la frecuencia se acerque a 0.5."

### Ejemplos reales

1. **Control de calidad:** La probabilidad de que una pieza sea defectuosa se estima como el número de defectuosas dividido por el total producido.
2. **Medicina:** La eficacia de un tratamiento se estima a partir de ensayos clínicos con muchos pacientes.
3. **Seguros:** Las primas se calculan basándose en el historial de siniestros.
4. **Meteorología:** "Probabilidad de lluvia del 30%" significa que en el 30% de días con condiciones similares ha llovido.
5. **Deportes:** El porcentaje de aciertos de un jugador de baloncesto es una probabilidad frecuentista.

### Comparación con Laplace

| Aspecto | Laplace | Frecuentista |
|---------|---------|--------------|
| **Fundamento** | Simetría y equiprobabilidad | Experimentación repetida |
| **Cálculo** | Casos favorables / casos posibles | Frecuencia observada |
| **Ventaja** | No requiere datos | Se adapta a la realidad |
| **Desventaja** | Solo aplica a espacios equiprobables | Requiere muchas repeticiones |

### Ejemplo resuelto

**Problema:** Una fábrica produce 1000 bombillas. Se prueban 100 y se encuentran 3 defectuosas. ¿Cuál es la probabilidad de que una bombilla sea defectuosa?

**Solución (frecuentista):**
- $P(defectuosa) ≈ 3/100 = 0.03$

Si se probaran más bombillas, la estimación sería más precisa.

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets


print("Kernel listo y configurado")
```

```python
# Probabilidad Frecuentista - Demostración interactiva


def probabilidad_frecuentista(p_teorica=0.5, n_experimentos=100):
    """
    Demostración de la probabilidad frecuentista
    """
    np.random.seed(42)
    
    # Simular experimentos
    resultados = np.random.random(n_experimentos) < p_teorica
    frecuencias_acum = np.cumsum(resultados) / np.arange(1, n_experimentos + 1)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: Convergencia de la frecuencia relativa
    axes[0].plot(range(1, n_experimentos + 1), frecuencias_acum, 'b-', linewidth=1, alpha=0.7)
    axes[0].axhline(y=p_teorica, color='red', linestyle='--', linewidth=2, label=f'Probabilidad teórica = {p_teorica}')
    axes[0].set_xlabel('Número de experimentos (n)')
    axes[0].set_ylabel('Frecuencia relativa')
    axes[0].set_title('Ley de los Grandes Números\nLa frecuencia se acerca a la probabilidad teórica')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    axes[0].set_ylim(0, 1)
    
    # Gráfico 2: Frecuencia final vs teórica
    freq_final = frecuencias_acum[-1]
    categorias = ['Frecuencia observada', 'Probabilidad teórica']
    valores = [freq_final, p_teorica]
    colores = ['skyblue', 'lightcoral']
    
    axes[1].bar(categorias, valores, color=colores, edgecolor='black')
    axes[1].set_ylabel('Valor')
    axes[1].set_title(f'n = {n_experimentos} experimentos\nFrecuencia = {freq_final:.4f}')
    axes[1].set_ylim(0, 1)
    axes[1].grid(axis='y', alpha=0.3)
    
    # Información
    error_abs = abs(freq_final - p_teorica)
    info = f"""
    Probabilidad Frecuentista
    
    • Probabilidad teórica (Laplace): P = {p_teorica}
    • Número de experimentos realizados: n = {n_experimentos}
    • Frecuencia observada: {freq_final:.4f}
    • Error absoluto: {error_abs:.6f}
    
    Interpretación: La ley de los grandes números garantiza que
       la frecuencia relativa se aproxima a la probabilidad teórica
       cuando el número de experimentos es grande.
    """
    
    plt.suptitle('Probabilidad Frecuentista', fontsize=14, fontweight='bold')
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()

interact(probabilidad_frecuentista,
         p_teorica=widgets.FloatSlider(min=0, max=1, step=0.05, value=0.5,
                                       description='P teórica:',
                                       style={'description_width': 'initial'},
                                       continuous_update=False),
         n_experimentos=widgets.IntSlider(min=10, max=5000, step=10, value=100,
                                          description='Experimentos (n):',
                                          style={'description_width': 'initial'},
                                          continuous_update=False))
```

## Probabilidad Geométrica (Problema del Encuentro)

### Definición

La probabilidad geométrica se aplica cuando el espacio muestral es un conjunto continuo (área, longitud, volumen). La probabilidad de un evento es la medida (longitud, área, volumen) del conjunto favorable dividida por la medida del espacio muestral.

$$P(A) = \frac{\text{Medida del evento favorable}}{\text{Medida del espacio muestral}}$$

### El problema clásico del encuentro

> "Dos personas acuerdan encontrarse en un lugar entre las 12:00 y las 13:00. Cada una llega en un momento aleatorio y espera 15 minutos. ¿Cuál es la probabilidad de que se encuentren?"

### Analogía

> "Imagina que el espacio muestral es un cuadrado de 1 hora × 1 hora (60×60 minutos). Las llegadas de cada persona son coordenadas (x, y). Se encuentran si la diferencia entre sus tiempos de llegada es menor o igual a 15 minutos."

### Ejemplos reales

1. **Transporte público:** Probabilidad de que dos personas tomen el mismo autobús si llegan en horarios aleatorios.
2. **Citas:** Probabilidad de que dos amigos se encuentren si cada uno espera un tiempo limitado.
3. **Llamadas telefónicas:** Probabilidad de que dos llamadas se solapen en el tiempo.
4. **Llegada a un servidor:** Probabilidad de que dos solicitudes lleguen a un servidor dentro de un intervalo de tiempo dado.
5. **Teoría de colas:** Probabilidad de que un cliente no tenga que esperar.

### Fórmula general (problema del encuentro)

Si dos personas llegan aleatoriamente en un intervalo de tiempo $T$ y cada una espera un tiempo $t$:

$$P(\text{encuentro}) = 1 - \left(1 - \frac{t}{T}\right)^2$$

Para el caso clásico: $T = 60$ minutos, $t = 15$ minutos:

$$P = 1 - \left(1 - \frac{15}{60}\right)^2 = 1 - \left(\frac{3}{4}\right)^2 = 1 - \frac{9}{16} = \frac{7}{16} = 0.4375$$

### Representación geométrica

El espacio muestral es un cuadrado de lado T. La región favorable es la franja donde $|x - y| ≤ t$.



El área favorable es: $T^2 - (T - t)^2$

### Ejemplo resuelto

**Problema:** Dos personas llegan entre las 8:00 y las 9:00. Cada una espera 10 minutos. ¿Probabilidad de encuentro?

**Solución:**
- $T = 60$ minutos
- $t = 10$ minutos
- $P = 1 - (1 - 10/60)^2 = 1 - (5/6)^2 = 1 - 25/36 = 11/36 ≈ 0.3056$

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets



print("Kernel listo y configurado")
```

```python
# Probabilidad Geométrica - Problema del Encuentro (VERSIÓN CORREGIDA)


def problema_encuentro(T=60, espera=15, n_simulaciones=1000):
    """
    Demostración del problema geométrico del encuentro
    """
    np.random.seed(42)
    
    # Cálculo teórico
    p_teorica = 1 - (1 - espera/T)**2
    
    # Simulación
    llegadas1 = np.random.uniform(0, T, n_simulaciones)
    llegadas2 = np.random.uniform(0, T, n_simulaciones)
    
    se_encuentran = np.abs(llegadas1 - llegadas2) <= espera
    p_simulada = np.mean(se_encuentran)
    
    # Crear figura con subgráficos
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # ============================================
    # GRÁFICO 1: Región de encuentro (teórica)
    # ============================================
    
    # Crear malla de puntos para visualizar la región
    # En lugar de fill_between con where, usamos una aproximación más simple
    
    # Mostrar el cuadrado completo
    axes[0].set_xlim(0, T)
    axes[0].set_ylim(0, T)
    axes[0].set_xlabel('Tiempo de llegada de la persona 1 (minutos)', fontsize=11)
    axes[0].set_ylabel('Tiempo de llegada de la persona 2 (minutos)', fontsize=11)
    axes[0].set_title('Región donde se encuentran (Teórica)', fontsize=12)
    axes[0].grid(True, alpha=0.3)
    
    # Dibujar el cuadrado de fondo
    axes[0].fill_between([0, T], 0, T, color='lightgray', alpha=0.3)
    
    # Dibujar la región de encuentro (franja donde |x-y| <= espera)
    # Método: crear un polígono que representa la región
    x_points = []
    y_points = []
    
    # Esquinas de la región favorable
    # Esquina superior izquierda de la franja
    x_points.append(0)
    y_points.append(espera)
    # Punto donde la línea x - y = espera corta el límite superior
    if espera <= T:
        x_points.append(T - espera)
        y_points.append(T)
    # Esquina superior derecha
    x_points.append(T)
    y_points.append(T)
    # Punto donde la línea y - x = espera corta el límite derecho
    if espera <= T:
        x_points.append(T)
        y_points.append(T - espera)
    # Esquina inferior derecha
    x_points.append(T)
    y_points.append(0)
    # Punto donde la línea y - x = espera corta el límite inferior
    if espera <= T:
        x_points.append(espera)
        y_points.append(0)
    
    # Para una visualización más clara, usamos una aproximación por puntos
    # Creamos una malla de puntos y coloreamos según la condición
    resolution = 100
    x_malla = np.linspace(0, T, resolution)
    y_malla = np.linspace(0, T, resolution)
    X, Y = np.meshgrid(x_malla, y_malla)
    
    # Máscara: |x-y| <= espera
    mascara = np.abs(X - Y) <= espera
    
    # Mostrar la región coloreada
    axes[0].imshow(mascara.T, origin='lower', extent=[0, T, 0, T], 
                   alpha=0.5, cmap='Blues', aspect='auto')
    
    # Línea x = y
    axes[0].plot([0, T], [0, T], 'r--', linewidth=2, label='x = y (llegada simultánea)')
    
    # Límites de la franja
    axes[0].plot([espera, T], [0, T - espera], 'g--', alpha=0.7, label=f'|x-y| = {espera} min')
    axes[0].plot([0, T - espera], [espera, T], 'g--', alpha=0.7)
    
    axes[0].legend(loc='upper left')
    
    # ============================================
    # GRÁFICO 2: Simulación de puntos
    # ============================================
    
    colores = ['green' if s else 'red' for s in se_encuentran[:500]]
    axes[1].scatter(llegadas1[:500], llegadas2[:500], c=colores, s=10, alpha=0.6)
    axes[1].set_xlabel('Tiempo de llegada de la persona 1 (minutos)', fontsize=11)
    axes[1].set_ylabel('Tiempo de llegada de la persona 2 (minutos)', fontsize=11)
    axes[1].set_title(f'Simulación (n={n_simulaciones})\nVerde = se encuentran, Rojo = no se encuentran', fontsize=12)
    axes[1].set_xlim(0, T)
    axes[1].set_ylim(0, T)
    axes[1].grid(True, alpha=0.3)
    
    # Añadir la línea x=y en la simulación
    axes[1].plot([0, T], [0, T], 'k--', alpha=0.5, linewidth=1.5)
    
    # Calcular áreas
    area_total = T * T
    area_favorable = T*T - (T - espera)*(T - espera)
    
    # Información
    info = f"""
    📊 PROBLEMA GEOMÉTRICO DEL ENCUENTRO
    
    Parámetros:
    • Intervalo de tiempo: T = {T} minutos
    • Tiempo de espera: t = {espera} minutos
    
    Cálculo teórico (geometría):
    • Área total: A_total = {area_total:.0f} min²
    • Área favorable (|x-y| ≤ t): A_favorable = {area_favorable:.1f} min²
    • P(encuentro) = A_favorable / A_total = {p_teorica:.4f}
    
    Simulación (frecuentista):
    • Número de simulaciones: {n_simulaciones}
    • Encuentros simulados: {np.sum(se_encuentran)}
    • P(encuentro) simulada = {p_simulada:.4f}
    • Error absoluto = {abs(p_simulada - p_teorica):.6f}
    
    Fórmula: P = 1 - (1 - t/T)²
    """
    
    plt.suptitle('Probabilidad Geométrica - Problema del Encuentro', fontsize=14, fontweight='bold')
    fig.text(0.5, -0.02, info, ha='center', fontsize=9,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()
    
    # Mostrar resultado en consola
    print("\n" + "="*50)
    print("📌 INTERPRETACIÓN DEL PROBLEMA DEL ENCUENTRO")
    print("="*50)
    print(f"""
    Si dos personas llegan aleatoriamente en un intervalo de {T} minutos,
    y cada una espera {espera} minutos, la probabilidad de que se encuentren es:
    
    • Teórica: {p_teorica:.2%}
    • Simulada: {p_simulada:.2%}
    
    Esto significa que en aproximadamente {int(p_teorica*100)} de cada 100 intentos,
    las dos personas logran encontrarse.
    """)
    print("="*50)


# ============================================
# INTERFAZ INTERACTIVA
# ============================================

interact(problema_encuentro,
         T=widgets.IntSlider(min=30, max=120, step=5, value=60,
                             description='Intervalo T (min):',
                             style={'description_width': 'initial'},
                             continuous_update=False),
         espera=widgets.IntSlider(min=5, max=30, step=1, value=15,
                                   description='Espera t (min):',
                                   style={'description_width': 'initial'},
                                   continuous_update=False),
         n_simulaciones=widgets.IntSlider(min=500, max=5000, step=500, value=1000,
                                          description='Simulaciones:',
                                          style={'description_width': 'initial'},
                                          continuous_update=False))
```

## Resumen: Tres enfoques de la probabilidad

| Enfoque | Fórmula | Cuándo usar | Ejemplo |
|---------|---------|-------------|---------|
| **Laplaciana (clásica)** | Casos favorables / Casos posibles | Resultados equiprobables y espacio finito | Lanzar un dado justo |
| **Frecuentista** | Límite de frecuencia relativa | Experimentos repetibles | Control de calidad |
| **Geométrica** | Medida favorable / Medida total | Espacios continuos | Problema del encuentro |
