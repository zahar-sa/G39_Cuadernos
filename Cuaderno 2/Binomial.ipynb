$$** Distribución Binomial **

$$Resumen



| Característica | Valor |
|---|---|
| **Notación** | $X \sim \text{Bernoulli}(p)$ |
| **Espacio muestral** | $\{0, 1\}$ (Fracaso, Éxito) |
| **Parámetros** | $p \in (0,1)$ (probabilidad de éxito) |
| **Función de masa (pmf)** | $P(X=1)=p,\; P(X=0)=1-p$ |
| **Esperanza** | $\mathbb{E}[X] = p$ |
| **Varianza** | $\text{Var}(X) = p(1-p)$ |
| **Usos típicos** | Una sola prueba con dos resultados (cara/sello, sí/no, éxito/fracaso) |

$$Programación

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

def graficar_binomial(n=20, p=0.5):
    x = np.arange(0, n+1)
    pmf = binom.pmf(x, n, p)
    plt.figure(figsize=(8,4))
    plt.bar(x, pmf, color='skyblue', edgecolor='black')
    plt.title(f'Distribución Binomial: n={n}, p={p}')
    plt.xlabel('Número de éxitos')
    plt.ylabel('Probabilidad')
    plt.ylim(0, max(pmf)*1.1)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Cambia los valores manualmente para ver diferentes gráficas
graficar_binomial(n=10, p=0.3)
graficar_binomial(n=20, p=0.5)
graficar_binomial(n=30, p=0.7)
```
