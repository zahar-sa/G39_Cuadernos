## Distribución Beta

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Beta}(\alpha, \beta)$ |
| **Parámetros** | $\alpha > 0$, $\beta > 0$ |
| **Soporte** | $[0, 1]$ |
| **Función de densidad** | $f(x) = \frac{x^{\alpha-1}(1-x)^{\beta-1}}{B(\alpha, \beta)}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{\alpha}{\alpha + \beta}$ |
| **Varianza** | $\text{Var}(X) = \frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$ |

### 🎯 Analogía

> "La Beta es la distribución de **probabilidades**. Modela la incertidumbre sobre una probabilidad desconocida."

### 📝 Ejemplos

1. **Probabilidad de éxito:** Incertidumbre sobre p en una binomial
2. **Ratings:** Calificaciones en plataformas (0-5 estrellas escalado a [0,1])
3. **Teorema de Bayes:** Es la distribución conjugada de la binomial

### 🔢 Casos particulares

- **α = β = 1**: Uniforme(0,1)
- **α = β > 1**: Simétrica con centro en 0.5
- **α < 1, β < 1**: Forma de U (valores extremos más probables)

```python
# Distribución Beta - Completa
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import beta
from ipywidgets import interact, widgets

def beta_completa(alpha=2.0, beta_param=2.0):
    """
    Demostración completa de la distribución Beta
    """
    x = np.linspace(0, 1, 1000)
    pdf = beta.pdf(x, alpha, beta_param)
    cdf = beta.cdf(x, alpha, beta_param)
    
    esperanza = alpha / (alpha + beta_param)
    varianza = (alpha * beta_param) / ((alpha + beta_param)**2 * (alpha + beta_param + 1))
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf, alpha=0.3, color='blue')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('Densidad f(x)')
    axes[0].set_title(f'Distribución Beta\nα = {alpha}, β = {beta_param}', fontsize=12)
    axes[0].axvline(x=esperanza, color='red', linestyle='--', label=f'Media = {esperanza:.3f}')
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    axes[0].set_ylim(0, max(pdf) * 1.1)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf, alpha=0.3, color='green')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('Probabilidad acumulada F(x)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • α = {alpha}, β = {beta_param}
    • Esperanza: μ = {esperanza:.4f}
    • Varianza: σ² = {varianza:.6f}
    • Moda: {max(0, (alpha-1)/(alpha+beta_param-2)) if alpha>1 and beta_param>1 else 0.5:.4f}
    """
    
    plt.suptitle('Distribución Beta', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()

interact(beta_completa,
         alpha=widgets.FloatSlider(min=0.1, max=10, step=0.1, value=2,
                                   description='α (alfa):'),
         beta_param=widgets.FloatSlider(min=0.1, max=10, step=0.1, value=2,
                                        description='β (beta):'))
```
