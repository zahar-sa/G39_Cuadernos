```python
## Distribución Bernoulli

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Bernoulli}(p)$ |
| **Parámetro** | $p \in (0,1)$ = probabilidad de éxito |
| **Espacio muestral** | $\{0, 1\}$ (Fracaso = 0, Éxito = 1) |
| **Función de masa** | $P(X=1)=p$, $P(X=0)=1-p$ |
| **Esperanza** | $\mathbb{E}[X] = p$ |
| **Varianza** | $\text{Var}(X) = p(1-p)$ |
| **Usos típicos** | Una sola prueba con dos resultados |

### 🪙 Analogía

> "Es como **lanzar una moneda una sola vez**, pero la moneda puede estar cargada. Si sale cara (éxito) ganas, si sale sello (fracaso) pierdes."

### 📝 Ejemplos

- **Control de calidad:** Una pieza sale defectuosa (1) o no (0)
- **Medicina:** Un paciente responde al tratamiento (1) o no (0)
- **Encuesta:** Una persona vota por el candidato A (1) o no (0)

### 🔗 Relación con Binomial

> La Bernoulli es el **caso base** de la Binomial. Si repites $n$ veces un experimento Bernoulli, obtienes una Binomial.

$$\text{Binomial}(n,p) = \sum_{i=1}^{n} \text{Bernoulli}_i(p)$$
```

```python
# Distribución Bernoulli - Completa
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli
from ipywidgets import interact, widgets

def bernoulli_completa(p=0.5, n_simulaciones=1000):
    """
    Demostración completa de la distribución Bernoulli
    """
    x = [0, 1]
    pmf = [1-p, p]
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Gráfico 1: PMF teórica
    axes[0].bar(x, pmf, width=0.4, color=['lightcoral', 'lightgreen'],
                edgecolor='black', alpha=0.8)
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(['Fracaso (0)', 'Éxito (1)'])
    axes[0].set_ylabel('Probabilidad')
    axes[0].set_title('Distribución Teórica', fontsize=12)
    axes[0].set_ylim(0, 1)
    axes[0].grid(axis='y', alpha=0.3)
    
    # Añadir valores
    for i, prob in enumerate(pmf):
        axes[0].text(i, prob + 0.02, f'{prob:.3f}', ha='center')
    
    # Gráfico 2: Simulación
    np.random.seed(42)
    simulacion = bernoulli.rvs(p, size=n_simulaciones)
    valores, conteos = np.unique(simulacion, return_counts=True)
    prob_sim = conteos / n_simulaciones
    
    axes[1].bar(valores, prob_sim, width=0.4, 
                color=['lightcoral', 'lightgreen'],
                edgecolor='black', alpha=0.8)
    axes[1].set_xticks(x)
    axes[1].set_xticklabels(['Fracaso (0)', 'Éxito (1)'])
    axes[1].set_ylabel('Frecuencia relativa')
    axes[1].set_title(f'Simulación (n={n_simulaciones})', fontsize=12)
    axes[1].set_ylim(0, 1)
    axes[1].grid(axis='y', alpha=0.3)
    
    # Gráfico 3: Proporción acumulada de éxitos
    exitos_acum = np.cumsum(simulacion)
    prop_acum = exitos_acum / np.arange(1, n_simulaciones + 1)
    
    axes[2].plot(range(1, n_simulaciones + 1), prop_acum, 
                 'b-', alpha=0.5, linewidth=0.5)
    axes[2].axhline(y=p, color='red', linestyle='--', 
                    label=f'p teórica = {p}')
    axes[2].set_xlabel('Número de ensayos')
    axes[2].set_ylabel('Proporción de éxitos')
    axes[2].set_title('Convergencia a la probabilidad teórica', fontsize=12)
    axes[2].set_ylim(0, 1)
    axes[2].legend()
    axes[2].grid(alpha=0.3)
    
    # Información
    esperanza = p
    varianza = p * (1 - p)
    
    info = f"""📊 Estadísticos:
    • Esperanza (teórica): μ = {esperanza:.4f}
    • Esperanza (simulada): μ̂ = {prop_acum[-1]:.4f}
    • Varianza (teórica): σ² = {varianza:.4f}
    • Error absoluto: {abs(prop_acum[-1] - p):.4f}
    """
    
    plt.suptitle('Distribución Bernoulli', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Probabilidades específicas
    print(f"\n📋 Probabilidades:")
    print(f"  P(Éxito) = P(X=1) = {p:.4f}")
    print(f"  P(Fracaso) = P(X=0) = {1-p:.4f}")

interact(bernoulli_completa,
         p=widgets.FloatSlider(min=0, max=1, step=0.01, value=0.5,
                               description='p (éxito):'),
         n_simulaciones=widgets.IntSlider(min=100, max=10000, step=100, 
                                          value=1000, description='Simulaciones:'))
```
