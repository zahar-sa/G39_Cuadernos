## Distribución Binomial Negativa

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{NB}(r, p)$ |
| **Parámetros** | $r$ = número de éxitos deseados, $p$ = probabilidad de éxito |
| **Espacio muestral** | $\{r, r+1, r+2, \dots\}$ |
| **Función de masa** | $P(X=k) = \binom{k-1}{r-1} p^r (1-p)^{k-r}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{r}{p}$ |
| **Varianza** | $\text{Var}(X) = \frac{r(1-p)}{p^2}$ |
| **Usos típicos** | Número de ensayos hasta obtener $r$ éxitos |

### 🎯 Analogía

> "Es como **seguir lanzando una moneda hasta obtener 3 caras**. Cuentas cuántos lanzamientos necesitas en total (incluyendo las 3 caras)."

### 📝 Ejemplos

1. **Control de calidad:** Número de inspecciones hasta encontrar 5 defectuosos
2. **Deportes:** Número de tiros hasta anotar 10 goles
3. **Medicina:** Número de pacientes hasta tener 20 que responden al tratamiento

### 🔗 Relación con otras distribuciones

- **Geométrica:** Es el caso particular cuando $r=1$
- **Binomial:** Cuenta éxitos en número fijo de ensayos (BN cuenta ensayos hasta número fijo de éxitos)

```python
# Distribución Binomial Negativa - Completa
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import nbinom
from ipywidgets import interact, widgets

def binomial_negativa_completa(r=3, p=0.4, max_k=30):
    """
    Demostración completa de la distribución Binomial Negativa
    """
    x = np.arange(r, max_k + 1)
    pmf = nbinom.pmf(x - r, r, p)  # Ajuste: nbinom.pmf(k) = P(X = r+k)
    
    esperanza = r / p
    varianza = r * (1 - p) / p**2
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PMF
    axes[0].bar(x, pmf, width=0.6, color='lightcoral', 
                edgecolor='black', alpha=0.7)
    axes[0].set_xlabel('Número total de ensayos (k)', fontsize=12)
    axes[0].set_ylabel('Probabilidad', fontsize=12)
    axes[0].set_title(f'Binomial Negativa: r={r}, p={p}', fontsize=12)
    axes[0].grid(axis='y', alpha=0.3)
    
    # Gráfico 2: CDF
    cdf = np.cumsum(pmf)
    axes[1].step(x, cdf, where='mid', color='darkred', linewidth=2)
    axes[1].fill_between(x, cdf, step='mid', alpha=0.3, color='lightcoral')
    axes[1].set_xlabel('Número total de ensayos (k)', fontsize=12)
    axes[1].set_ylabel('Probabilidad acumulada', fontsize=12)
    axes[1].set_title('Función de Distribución Acumulada', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • Esperanza: μ = {esperanza:.2f}
    • Varianza: σ² = {varianza:.4f}
    • Desviación: σ = {np.sqrt(varianza):.4f}
    • Probabilidad de éxito: p = {p}
    • Éxitos requeridos: r = {r}
    """
    
    plt.suptitle('Distribución Binomial Negativa', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Interpretación
    print(f"\n📋 Interpretación:")
    print(f"  Probabilidad de necesitar exactamente {r} ensayos: {pmf[0]:.4f}")
    print(f"  Probabilidad de necesitar más de {max_k} ensayos: {1 - cdf[-1]:.4f}")

interact(binomial_negativa_completa,
         r=widgets.IntSlider(min=1, max=20, step=1, value=3,
                             description='r (éxitos):'),
         p=widgets.FloatSlider(min=0.01, max=0.99, step=0.01, value=0.4,
                               description='p (éxito):'),
         max_k=widgets.IntSlider(min=10, max=100, step=5, value=30,
                                 description='Máximo k:'))
```
