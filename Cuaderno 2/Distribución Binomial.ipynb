## Distribución Binomial

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Binomial}(n, p)$ |
| **Parámetros** | $n$ = número de ensayos, $p$ = probabilidad de éxito |
| **Espacio muestral** | $\{0, 1, 2, \dots, n\}$ |
| **Función de masa** | $P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}$ |
| **Esperanza** | $\mathbb{E}[X] = np$ |
| **Varianza** | $\text{Var}(X) = np(1-p)$ |
| **Usos típicos** | Número de éxitos en $n$ ensayos independientes |

### 🎯 Analogía

> "Es como **lanzar una moneda varias veces** y contar cuántas caras salen. Cada lanzamiento es independiente y tiene la misma probabilidad de éxito."

### 📝 Ejemplos

1. **Control de calidad:** Número de piezas defectuosas en un lote de 100
2. **Medicina:** Número de pacientes que responden a un tratamiento de 50
3. **Encuestas:** Número de personas que votarán por un candidato en una muestra de 500

### 🔢 Propiedades importantes

1. **Aditividad:** Si $X \sim \text{Bin}(n_1, p)$ y $Y \sim \text{Bin}(n_2, p)$ independientes, entonces $X+Y \sim \text{Bin}(n_1+n_2, p)$

2. **Relación con Bernoulli:** 
   $$X = \sum_{i=1}^{n} B_i \quad \text{donde } B_i \sim \text{Bernoulli}(p)$$

3. **Aproximación a Poisson:** Si $n$ es grande y $p$ pequeño, $\text{Bin}(n,p) \approx \text{Poisson}(\lambda = np)$

```python
# Distribución Binomial - Completa
from scipy.stats import binom
from ipywidgets import interact, widgets

def binomial_completa(n=20, p=0.5, mostrar_poisson=False):
    """
    Demostración completa de la distribución Binomial
    """
    x = np.arange(0, n + 1)
    pmf_binom = binom.pmf(x, n, p)
    esperanza = n * p
    varianza = n * p * (1 - p)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PMF Binomial
    axes[0].bar(x, pmf_binom, width=0.8, color='skyblue', 
                edgecolor='black', alpha=0.7)
    axes[0].set_xlabel('Número de éxitos (k)', fontsize=12)
    axes[0].set_ylabel('Probabilidad', fontsize=12)
    axes[0].set_title(f'Binomial: n={n}, p={p}', fontsize=12)
    axes[0].set_ylim(0, max(pmf_binom) * 1.1)
    axes[0].grid(axis='y', alpha=0.3)
    
    # Añadir aproximación de Poisson si se solicita
    if mostrar_poisson and n > 30 and p < 0.1:
        from scipy.stats import poisson
        lam = n * p
        pmf_pois = poisson.pmf(x, lam)
        axes[0].plot(x, pmf_pois, 'ro-', markersize=4, 
                     label=f'Poisson(λ={lam:.2f})')
        axes[0].legend()
    
    # Gráfico 2: CDF
    cdf = binom.cdf(x, n, p)
    axes[1].step(x, cdf, where='mid', color='steelblue', linewidth=2)
    axes[1].fill_between(x, cdf, step='mid', alpha=0.3, color='steelblue')
    axes[1].set_xlabel('Número de éxitos (k)', fontsize=12)
    axes[1].set_ylabel('Probabilidad acumulada', fontsize=12)
    axes[1].set_title('Función de Distribución Acumulada', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • Esperanza: μ = {esperanza:.4f}
    • Varianza: σ² = {varianza:.4f}
    • Desviación: σ = {np.sqrt(varianza):.4f}
    • Moda: {int((n+1)*p)} (aproximadamente)
    • Asimetría: {(1-2*p)/np.sqrt(n*p*(1-p)):.4f}
    """
    
    plt.suptitle('Distribución Binomial', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Calcular probabilidades específicas
    print("\n📋 Algunas probabilidades:")
    print(f"  P(X = {n}) = {binom.pmf(n, n, p):.6f}")
    print(f"  P(X ≤ {n//2}) = {binom.cdf(n//2, n, p):.4f}")
    print(f"  P(X ≥ {n//2}) = {1 - binom.cdf(n//2 - 1, n, p):.4f}")

interact(binomial_completa,
         n=widgets.IntSlider(min=1, max=100, step=1, value=20,
                             description='n (ensayos):'),
         p=widgets.FloatSlider(min=0, max=1, step=0.01, value=0.5,
                               description='p (éxito):'),
         mostrar_poisson=widgets.Checkbox(value=False, 
                                          description='Mostrar aprox. Poisson'))
```
