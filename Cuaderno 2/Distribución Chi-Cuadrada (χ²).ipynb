## Distribución Chi-Cuadrada (χ²)

###  Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \chi^2_k$ |
| **Parámetro** | $k > 0$ = grados de libertad |
| **Soporte** | $[0, \infty)$ |
| **Función de densidad** | $f(x) = \frac{x^{k/2-1} e^{-x/2}}{2^{k/2} \Gamma(k/2)}$ |
| **Esperanza** | $\mathbb{E}[X] = k$ |
| **Varianza** | $\text{Var}(X) = 2k$ |

###  Analogía

> "La χ² es la **suma de k normales estándar al cuadrado**. Mide la 'distancia al cuadrado' entre lo observado y lo esperado."

###  Usos principales

1. **Prueba de independencia:** Tablas de contingencia
2. **Prueba de bondad de ajuste:** ¿Los datos siguen una distribución?
3. **Prueba de homogeneidad:** ¿Dos grupos son similares?
4. **Intervalos de confianza para la varianza**

###  Relación con otras distribuciones

- **χ² con 2 grados de libertad** = Exponencial(λ=1/2)
- **χ² con k grados** = Gamma(α=k/2, β=1/2)
- **Relación con Normal:** $Z \sim N(0,1) \Rightarrow Z^2 \sim \chi^2_1$

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias 

%matplotlib inline
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import chi2
from ipywidgets import interact, widgets

print("Kernel listo y configurado")
```

```python
# Distribución Chi-Cuadrada - Completa


def chicuadrada_completa(k=5):
    """
    Demostración completa de la distribución Chi-Cuadrada
    """
    x = np.linspace(0, chi2.ppf(0.995, k), 1000)
    pdf_chi2 = chi2.pdf(x, k)
    cdf_chi2 = chi2.cdf(x, k)
    
    esperanza = k
    varianza = 2 * k
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf_chi2, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf_chi2, alpha=0.3, color='blue')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('Densidad f(x)')
    axes[0].set_title(f'Distribución χ²\n{k} grados de libertad', fontsize=12)
    axes[0].axvline(x=esperanza, color='red', linestyle='--', label=f'Media = {esperanza}')
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf_chi2, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf_chi2, alpha=0.3, color='green')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('Probabilidad acumulada F(x)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • Grados de libertad: k = {k}
    • Esperanza: μ = {esperanza}
    • Varianza: σ² = {varianza}
    • Desviación: σ = {np.sqrt(varianza):.4f}
    • Moda: {max(0, k-2):.4f}
    """
    
    plt.suptitle('Distribución Chi-Cuadrada (χ²)', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Cuantiles importantes
    print(f"\n📋 Cuantiles importantes:")
    for p in [0.95, 0.975, 0.99]:
        print(f"  χ²_{p*100:.1f}% = {chi2.ppf(p, k):.4f}")

interact(chicuadrada_completa,
         k=widgets.IntSlider(min=1, max=50, step=1, value=5,
                             description='k (grados):'))
```

### 5 Aplicaciones Reales de la Distribución Chi-Cuadrada (χ²)

1.  **Pruebas de Independencia:** Analizar si existe una relación significativa entre dos variables categóricas, como la preferencia de voto (candidato A/B) y el género (hombre/mujer) en una encuesta electoral.

2.  **Pruebas de Bondad de Ajuste:** Verificar si los datos observados se ajustan a una distribución teórica específica, por ejemplo, comprobar si los números que salen en una ruleta siguen una distribución uniforme.

3.  **Genética Poblacional:** Evaluar si las proporciones de diferentes rasgos hereditarios en la descendencia se ajustan a las proporciones esperadas por las leyes de Mendel (ej. 9:3:3:1).

4.  **Control de Calidad (Atributos):** Determinar si la tasa de defectos de un producto es consistente a lo largo del tiempo comparando la frecuencia de unidades defectuosas en diferentes lotes de producción.

5.  **Análisis de Encuestas:** Evaluar si hay una asociación entre dos factores demográficos, como verificar si el nivel de educación (básico, medio, superior) está relacionado con la preferencia por un medio de comunicación específico (TV, radio, internet).
