## Distribución Exponencial

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Exp}(\lambda)$ |
| **Parámetro** | $\lambda > 0$ = tasa (eventos por unidad de tiempo) |
| **Soporte** | $[0, \infty)$ |
| **Función de densidad** | $f(x) = \lambda e^{-\lambda x}$ para $x \geq 0$ |
| **Función de distribución** | $F(x) = 1 - e^{-\lambda x}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{1}{\lambda}$ |
| **Varianza** | $\text{Var}(X) = \frac{1}{\lambda^2}$ |

### 🎯 Analogía

> "Modela el **tiempo de espera** hasta que ocurre un evento, como el tiempo hasta que recibes tu próximo email o el tiempo entre llegadas de autobuses."

### 📝 Ejemplos

1. **Tiempo de vida:** Vida útil de componentes electrónicos
2. **Llamadas telefónicas:** Tiempo entre llamadas a un call center
3. **Radiactividad:** Tiempo entre desintegraciones
4. **Servicio:** Tiempo de atención al cliente

### 🔑 Propiedad de Falta de Memoria

La distribución Exponencial es la **única** distribución continua con esta propiedad:

$$P(X > s + t \mid X > s) = P(X > t)$$

**Interpretación:** Si un componente ha durado $s$ horas, la probabilidad de que dure $t$ horas más es la misma que si fuera nuevo.

```python
# Distribución Exponencial - Completa
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import expon
from ipywidgets import interact, widgets

def exponencial_completa(lamb=1.0, mostrar_memoria=True):
    """
    Demostración completa de la distribución Exponencial
    """
    x = np.linspace(0, 5/lamb, 1000)
    pdf = expon.pdf(x, scale=1/lamb)
    cdf = expon.cdf(x, scale=1/lamb)
    
    esperanza = 1/lamb
    varianza = 1/lamb**2
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf, alpha=0.3, color='blue')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('Densidad f(x)')
    axes[0].set_title(f'Distribución Exponencial\nλ = {lamb}, E[X] = {esperanza:.3f}', fontsize=12)
    axes[0].axvline(x=esperanza, color='red', linestyle='--', label=f'Media = {esperanza:.3f}')
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf, alpha=0.3, color='green')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('Probabilidad acumulada F(x)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • Tasa: λ = {lamb}
    • Esperanza: μ = {esperanza:.4f}
    • Varianza: σ² = {varianza:.4f}
    • Desviación: σ = {np.sqrt(varianza):.4f}
    • Mediana: {np.log(2)/lamb:.4f}
    """
    
    if mostrar_memoria:
        s = esperanza
        t = esperanza
        p1 = 1 - expon.cdf(s + t, scale=1/lamb)
        p2 = 1 - expon.cdf(t, scale=1/lamb)
        p_cond = p1 / p2 if p2 > 0 else 0
        info += f"""
    📌 Propiedad de falta de memoria:
    • P(X > {s:.2f}+{t:.2f} | X > {s:.2f}) = {p_cond:.4f}
    • P(X > {t:.2f}) = {p2:.4f}
    • (Deben ser iguales)"""
    
    plt.suptitle('Distribución Exponencial', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    print(f"\n📋 Probabilidades de ejemplo:")
    print(f"  P(X ≤ {esperanza:.3f}) = {expon.cdf(esperanza, scale=1/lamb):.4f}")
    print(f"  P(X > {2*esperanza:.3f}) = {1 - expon.cdf(2*esperanza, scale=1/lamb):.4f}")

interact(exponencial_completa,
         lamb=widgets.FloatSlider(min=0.1, max=5, step=0.1, value=1,
                                  description='Tasa (λ):'),
         mostrar_memoria=widgets.Checkbox(value=True,
                                         description='Mostrar propiedad de memoria'))
```
