## Distribución F de Snedecor

###  Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $F \sim F_{d1, d2}$ |
| **Parámetros** | $d_1 > 0$, $d_2 > 0$ = grados de libertad |
| **Soporte** | $[0, \infty)$ |
| **Función de densidad** | $f(x) = \frac{\sqrt{\frac{(d_1 x)^{d_1} d_2^{d_2}}{(d_1 x + d_2)^{d_1+d_2}}}}{x B\left(\frac{d_1}{2}, \frac{d_2}{2}\right)}$ |
| **Esperanza** | $\mathbb{E}[F] = \frac{d_2}{d_2-2}$ (para $d_2 > 2$) |
| **Varianza** | $\text{Var}(F) = \frac{2d_2^2(d_1+d_2-2)}{d_1(d_2-2)^2(d_2-4)}$ (para $d_2 > 4$) |

###  Analogía

> "La F es el **cociente de dos varianzas** independientes. Se usa para comparar si dos grupos tienen la misma variabilidad."

###  Usos principales

1. **ANOVA:** Comparación de múltiples medias
2. **Prueba de igualdad de varianzas**
3. **Regresión:** Prueba de significancia global
4. **Comparación de modelos**

###  Relación con t de Student

$$F_{1, d_2} = t_{d_2}^2$$

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import f
from ipywidgets import interact, widgets

print("Kernel listo y configurado")
```

```python
# Distribución F - Completa


def f_snedecor_completa(d1=5, d2=10):
    """
    Demostración completa de la distribución F
    """
    x = np.linspace(0, f.ppf(0.99, d1, d2), 1000)
    pdf_f = f.pdf(x, d1, d2)
    cdf_f = f.cdf(x, d1, d2)
    
    esperanza = d2/(d2-2) if d2 > 2 else "No definida"
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf_f, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf_f, alpha=0.3, color='blue')
    axes[0].set_xlabel('F')
    axes[0].set_ylabel('Densidad f(F)')
    axes[0].set_title(f'Distribución F de Snedecor\nd₁ = {d1}, d₂ = {d2}', fontsize=12)
    axes[0].grid(alpha=0.3)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf_f, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf_f, alpha=0.3, color='green')
    axes[1].set_xlabel('F')
    axes[1].set_ylabel('Probabilidad acumulada F(F)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f""" Estadísticos:
    • Grados libertad numerador: d₁ = {d1}
    • Grados libertad denominador: d₂ = {d2}
    • Esperanza: μ = {esperanza}
    • Moda: {max(0, (d1-2)/d1 * d2/(d2+2)):.4f}
    """
    
    plt.suptitle('Distribución F de Snedecor', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Cuantiles importantes
    print(f"\n Cuantiles importantes (α = 0.05):")
    print(f"  F₀.₉₅ = {f.ppf(0.95, d1, d2):.4f}")
    print(f"  F₀.₉₉ = {f.ppf(0.99, d1, d2):.4f}")

interact(f_snedecor_completa,
         d1=widgets.IntSlider(min=1, max=50, step=1, value=5,
                              description='d₁ (numerador):'),
         d2=widgets.IntSlider(min=1, max=50, step=1, value=10,
                              description='d₂ (denominador):'))
```

### 5 Aplicaciones Reales de la Distribución F de Snedecor

1.  **Análisis de Varianza (ANOVA):** Comparar las medias de tres o más grupos (ej. eficacia de diferentes métodos de enseñanza) para determinar si al menos uno es significativamente diferente.

2.  **Comparación de Varianzas (Prueba F):** Verificar si dos poblaciones tienen la misma variabilidad (ej. comparar la consistencia de dos máquinas embotelladoras).

3.  **Regresión Lineal Múltiple:** Probar la significancia global del modelo, determinando si las variables independientes (ej. precio, publicidad) explican colectivamente la variable dependiente (ej. ventas).

4.  **Control de Calidad (Gráficos de Control):** Monitorear si la variabilidad de un proceso se mantiene dentro de límites aceptables comparando varianzas entre lotes de producción.

5.  **Ecología:** Evaluar si la diversidad de especies (varianza) es similar entre diferentes ecosistemas o áreas de estudio.
