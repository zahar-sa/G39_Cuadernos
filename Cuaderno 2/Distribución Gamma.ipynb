## Distribución Gamma

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Gamma}(\alpha, \beta)$ |
| **Parámetros** | $\alpha > 0$ = forma, $\beta > 0$ = tasa |
| **Soporte** | $[0, \infty)$ |
| **Función de densidad** | $f(x) = \frac{\beta^\alpha}{\Gamma(\alpha)} x^{\alpha-1} e^{-\beta x}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{\alpha}{\beta}$ |
| **Varianza** | $\text{Var}(X) = \frac{\alpha}{\beta^2}$ |

### 🎯 Analogía

> "La Gamma generaliza la Exponencial. Si la Exponencial es el tiempo hasta **un** evento, la Gamma es el tiempo hasta **α** eventos."

### 📝 Casos particulares

- **α = 1**: Exponencial(β)
- **α = n/2, β = 1/2**: χ² (Chi-cuadrado con n grados de libertad)
- **α entero**: Erlang (suma de α exponenciales independientes)

### 📝 Ejemplos

1. **Tiempo de espera:** Tiempo hasta que ocurren α eventos
2. **Seguros:** Modelar siniestros
3. **Hidrología:** Caudales de ríos

```python
# Distribución Gamma - Completa
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import gamma
from ipywidgets import interact, widgets

def gamma_completa(alpha=2.0, beta=1.0):
    """
    Demostración completa de la distribución Gamma
    """
    x = np.linspace(0, gamma.ppf(0.999, alpha, scale=1/beta), 1000)
    pdf = gamma.pdf(x, alpha, scale=1/beta)
    cdf = gamma.cdf(x, alpha, scale=1/beta)
    
    esperanza = alpha / beta
    varianza = alpha / beta**2
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf, alpha=0.3, color='blue')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('Densidad f(x)')
    axes[0].set_title(f'Distribución Gamma\nα = {alpha}, β = {beta}', fontsize=12)
    axes[0].axvline(x=esperanza, color='red', linestyle='--', label=f'Media = {esperanza:.3f}')
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf, alpha=0.3, color='green')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('Probabilidad acumulada F(x)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • Forma: α = {alpha}
    • Tasa: β = {beta}
    • Escala: θ = 1/β = {1/beta:.4f}
    • Esperanza: μ = {esperanza:.4f}
    • Varianza: σ² = {varianza:.4f}
    • Moda: {max(0, (alpha-1)/beta):.4f}
    """
    
    plt.suptitle('Distribución Gamma', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Comparación con Exponencial
    if alpha == 1:
        print("\n📌 Nota: α = 1 → Esta es una distribución Exponencial(β)")

interact(gamma_completa,
         alpha=widgets.FloatSlider(min=0.1, max=10, step=0.1, value=2,
                                   description='Forma (α):'),
         beta=widgets.FloatSlider(min=0.1, max=5, step=0.1, value=1,
                                  description='Tasa (β):'))
```
