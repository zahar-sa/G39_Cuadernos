## Distribución Hipergeométrica

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Hiper}(N, K, n)$ |
| **Parámetros** | $N$ = población total, $K$ = éxitos en población, $n$ = tamaño muestra |
| **Espacio muestral** | $\{\max(0, n-(N-K)), \dots, \min(n, K)\}$ |
| **Función de masa** | $P(X=k) = \frac{\binom{K}{k} \binom{N-K}{n-k}}{\binom{N}{n}}$ |
| **Esperanza** | $\mathbb{E}[X] = n \cdot \frac{K}{N}$ |
| **Varianza** | $\text{Var}(X) = n \cdot \frac{K}{N} \cdot \frac{N-K}{N} \cdot \frac{N-n}{N-1}$ |
| **Usos típicos** | Muestreo sin reemplazo, control de calidad |

### 🎯 Analogía

> "Es como **sacar bolas de una urna SIN devolverlas**. Si quieres saber cuántas bolas rojas sacarás al extraer 10 bolas de una urna que tiene 30 rojas y 20 azules, usas Hipergeométrica."

### 📝 Ejemplos

1. **Control de calidad:** Número de piezas defectuosas en una muestra sin reemplazo
2. **Lotería:** Número de aciertos al comprar un billete
3. **Ecología:** Número de animales marcados capturados en una segunda muestra

### 🔗 Comparación con Binomial

| Aspecto | Binomial | Hipergeométrica |
|---------|----------|-----------------|
| **Muestreo** | Con reemplazo | Sin reemplazo |
| **Independencia** | Ensayos independientes | Ensayos dependientes |
| **Probabilidad** | Constante $p$ | Cambia con cada extracción |
| **Aproximación** | - | Si $n \ll N$, ≈ Binomial |

```python
# Distribución Hipergeométrica - Completa
from scipy.stats import hypergeom
from ipywidgets import interact, widgets

def hipergeometrica_completa(N=50, K=20, n=10):
    """
    Demostración completa de la distribución Hipergeométrica
    """
    # Rango posible de valores
    k_min = max(0, n - (N - K))
    k_max = min(n, K)
    x = np.arange(k_min, k_max + 1)
    pmf = hypergeom.pmf(x, N, K, n)
    
    esperanza = n * K / N
    varianza = n * (K/N) * ((N-K)/N) * ((N-n)/(N-1))
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PMF
    axes[0].bar(x, pmf, width=0.6, color='plum', 
                edgecolor='black', alpha=0.7)
    axes[0].set_xlabel('Número de éxitos en la muestra (k)', fontsize=12)
    axes[0].set_ylabel('Probabilidad', fontsize=12)
    axes[0].set_title
```
