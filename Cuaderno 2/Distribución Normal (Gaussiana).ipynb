## Distribución Normal (Gaussiana)

###  Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim N(\mu, \sigma^2)$ |
| **Parámetros** | $\mu$ = media, $\sigma^2$ = varianza, $\sigma$ = desviación |
| **Soporte** | $(-\infty, \infty)$ |
| **Función de densidad** | $f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}$ |
| **Función de distribución** | No tiene forma cerrada (se usa tabla) |
| **Esperanza** | $\mathbb{E}[X] = \mu$ |
| **Varianza** | $\text{Var}(X) = \sigma^2$ |

###  Analogía

> "Es la **distribución de los errores** en mediciones. Si mides muchas veces algo, los errores tienden a seguir una campana de Gauss."

###  Ejemplos

1. **Altura:** Distribución de alturas en una población
2. **CI:** Puntuaciones de coeficiente intelectual
3. **Errores de medición:** Errores aleatorios en instrumentos
4. **Rendimiento académico:** Calificaciones en un examen

###  Regla Empírica (68-95-99.7)

- **68%** de los datos están en $\mu \pm \sigma$
- **95%** de los datos están en $\mu \pm 2\sigma$
- **99.7%** de los datos están en $\mu \pm 3\sigma$

###  Teorema Central del Límite (TCL)

> La suma de muchas variables aleatorias independientes (con varianza finita) se aproxima a una distribución Normal.

$$\bar{X}_n \xrightarrow{d} N\left(\mu, \frac{\sigma^2}{n}\right)$$

```python
%matplotlib inline
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import norm
from ipywidgets import interact, widgets

print("Kernel listo y configurado")
```

```python
# Distribución Normal - Completa


def normal_completa(mu=0, sigma=1, mostrar_regla_empirica=True):
    """
    Demostración completa de la distribución Normal
    """
    x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
    pdf = norm.pdf(x, mu, sigma)
    cdf = norm.cdf(x, mu, sigma)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf, alpha=0.3, color='blue')
    axes[0].axvline(x=mu, color='red', linestyle='-', linewidth=2, label=f'μ = {mu}')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('Densidad f(x)')
    axes[0].set_title(f'Distribución Normal\nN({mu}, {sigma}²)', fontsize=12)
    axes[0].grid(alpha=0.3)
    
    # Añadir regla empírica
    if mostrar_regla_empirica:
        # 1 sigma
        axes[0].axvline(x=mu - sigma, color='green', linestyle='--', alpha=0.5)
        axes[0].axvline(x=mu + sigma, color='green', linestyle='--', alpha=0.5)
        # 2 sigma
        axes[0].axvline(x=mu - 2*sigma, color='orange', linestyle='--', alpha=0.5)
        axes[0].axvline(x=mu + 2*sigma, color='orange', linestyle='--', alpha=0.5)
    
    axes[0].legend()
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf, alpha=0.3, color='green')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('Probabilidad acumulada F(x)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Estadísticos:
    • Media: μ = {mu}
    • Varianza: σ² = {sigma**2}
    • Desviación: σ = {sigma}
    • Asimetría: 0
    • Curtosis: 3 (exceso = 0)
    """
    
    if mostrar_regla_empirica:
        info += f"""
    📌 Regla Empírica (68-95-99.7):
    • P(|X-μ| < σ) = {norm.cdf(mu+sigma, mu, sigma) - norm.cdf(mu-sigma, mu, sigma):.4f}
    • P(|X-μ| < 2σ) = {norm.cdf(mu+2*sigma, mu, sigma) - norm.cdf(mu-2*sigma, mu, sigma):.4f}
    • P(|X-μ| < 3σ) = {norm.cdf(mu+3*sigma, mu, sigma) - norm.cdf(mu-3*sigma, mu, sigma):.4f}
    """
    
    plt.suptitle('Distribución Normal (Gaussiana)', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Cuantiles importantes
    print(f"\n📋 Cuantiles importantes:")
    for p in [0.025, 0.16, 0.5, 0.84, 0.975]:
        print(f"  Percentil {p*100:5.1f}%: {norm.ppf(p, mu, sigma):.4f}")

interact(normal_completa,
         mu=widgets.FloatSlider(min=-5, max=5, step=0.1, value=0,
                                description='Media (μ):'),
         sigma=widgets.FloatSlider(min=0.1, max=3, step=0.1, value=1,
                                   description='Desv. (σ):'),
         mostrar_regla_empirica=widgets.Checkbox(value=True,
                                                description='Mostrar regla 68-95-99.7'))
```

### 5 Aplicaciones Reales de la Distribución Normal

1.  **Altura de personas:** La estatura de una población (hombres o mujeres) sigue aproximadamente una distribución normal, permitiendo calcular porcentajes de personas por rangos de altura.

2.  **Calificaciones educativas:** En exámenes estandarizados (como el SAT, GRE o pruebas de acceso a la universidad), las puntuaciones se ajustan para seguir una curva normal (campana de Gauss).

3.  **Control de calidad industrial:** Las dimensiones de productos fabricados en masa (diámetro de un eje, grosor de una lámina) se distribuyen normalmente alrededor de la especificación objetivo.

4.  **Errores de medición:** En experimentos científicos, los errores aleatorios al medir una magnitud (temperatura, peso, longitud) siguen una distribución normal.

5.  **Presión arterial:** Los niveles de presión arterial sistólica en una población sana suelen aproximarse a una distribución normal para cada grupo etario y género.
