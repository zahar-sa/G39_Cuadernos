## Distribución Uniforme Continua

###  Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim U(a, b)$ |
| **Parámetros** | $a$ = mínimo, $b$ = máximo |
| **Soporte** | $[a, b]$ |
| **Función de densidad** | $f(x) = \frac{1}{b-a}$ para $a \leq x \leq b$ |
| **Función de distribución** | $F(x) = \frac{x-a}{b-a}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{a+b}{2}$ |
| **Varianza** | $\text{Var}(X) = \frac{(b-a)^2}{12}$ |

###  Analogía

> "Es como **elegir un punto al azar** en una línea recta entre $a$ y $b$. Todos los puntos tienen la misma probabilidad de ser elegidos."

###  Ejemplos

1. **Tiempo de espera:** Llegas a una parada de autobús que pasa cada 10 minutos
2. **Números aleatorios:** Generar números entre 0 y 1
3. **Precisión de medición:** Error uniforme en un instrumento

###  Propiedades importantes

- Es el **análogo continuo** de la uniforme discreta
- Es la distribución con **máxima entropía** para un intervalo acotado
- Si $U \sim U(0,1)$, entonces $a + (b-a)U \sim U(a,b)$

### 5 Aplicaciones Reales de la Distribución Uniforme Continua

1.  **Generación de Números Aleatorios:** Es la base para generar números pseudoaleatorios en un rango [0,1), fundamentales para simulaciones y muestreo en software y videojuegos.

2.  **Llegada de Autobuses:** Modelar el tiempo de espera de un pasajero que llega a una parada sin saber el horario, si un autobús pasa cada intervalo fijo (ej. cada 10 minutos).

3.  **Industria (Corte de Materiales):** Modelar el punto de corte a lo largo de un rollo de tela, papel o cable, cuando no hay preferencia por un punto específico del material.

4.  **Teoría de Colas (Redondeo):** En sistemas muy congestionados o de alta velocidad, el tiempo de procesamiento de una tarea se puede aproximar a una distribución uniforme.

5.  **Muestreo Ambiental:** Seleccionar puntos aleatorios de muestreo dentro de un área geográfica cuadrada o rectangular, donde cualquier punto tiene la misma probabilidad de ser elegido.
