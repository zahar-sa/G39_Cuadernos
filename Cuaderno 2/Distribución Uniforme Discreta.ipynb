## Distribución Uniforme Discreta

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Unif}\{a, a+1, \dots, b\}$ |
| **Parámetros** | $a$ (mínimo), $b$ (máximo), $n = b-a+1$ |
| **Espacio muestral** | $\{a, a+1, \dots, b\}$ |
| **Función de masa (pmf)** | $P(X=k) = \frac{1}{n}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{a+b}{2}$ |
| **Varianza** | $\text{Var}(X) = \frac{n^2 - 1}{12}$ |
| **Usos típicos** | Dados, rifas, selección aleatoria simple |

### 🎲 Analogía

> "Es como una **rifa justa**: todos los boletos tienen exactamente la misma probabilidad de ganar. No hay favoritismo, todos son iguales."

### 📝 Ejemplos

1. **Dado justo:** $X \sim \text{Unif}\{1,6\}$, $P(X=k) = 1/6$
2. **Rifa del 00 al 99:** $X \sim \text{Unif}\{0,99\}$, $P(X=k) = 1/100$
3. **Seleccionar un día de la semana:** $X \sim \text{Unif}\{1,7\}$

### 🔢 Demostración de la Esperanza

Para $X \sim \text{Unif}\{a, b\}$:

$$\mathbb{E}[X] = \sum_{k=a}^{b} k \cdot \frac{1}{n} = \frac{1}{n} \sum_{k=a}^{b} k = \frac{1}{n} \cdot \frac{n(a+b)}{2} = \frac{a+b}{2}$$

Donde usamos la fórmula: $\sum_{k=a}^{b} k = \frac{n(a+b)}{2}$ con $n = b-a+1$.

# Distribución Uniforme Discreta - Completa
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint
from ipywidgets import interact, widgets

def uniforme_completa(a=1, b=6, mostrar_cdf=True):
    """
    Demostración completa de la distribución uniforme discreta
    """
    x = np.arange(a, b + 1)
    n = len(x)
    pmf = np.full_like(x, 1/n)
    esperanza = (a + b) / 2
    varianza = ((b - a + 1)**2 - 1) / 12
    
    fig, axes = plt.subplots(1, 2 if mostrar_cdf else 1, 
                              figsize=(14 if mostrar_cdf else 8, 5))
    
    if not mostrar_cdf:
        axes = [axes]
    
    # Gráfico 1: PMF
    axes[0].bar(x, pmf, width=0.6, color='coral', 
                edgecolor='black', alpha=0.8)
    axes[0].set_xlabel('Valores', fontsize=12)
    axes[0].set_ylabel('Probabilidad', fontsize=12)
    axes[0].set_title(f'Función de Masa (PMF)\nUnif{{{a}, {b}}}', fontsize=12)
    axes[0].set_ylim(0, max(pmf) * 1.2)
    axes[0].grid(axis='y', alpha=0.3)
    
    # Añadir valores sobre las barras
    for i, (val, prob) in enumerate(zip(x, pmf)):
        axes[0].text(val, prob + 0.005, f'{prob:.3f}', 
                    ha='center', fontsize=9)
    
    # Gráfico 2: CDF (si se solicita)
    if mostrar_cdf:
        cdf = np.cumsum(pmf)
        axes[1].step(x, cdf, where='mid', color='steelblue', 
                     linewidth=2)
        axes[1].fill_between(x, cdf, step='mid', alpha=0.3, 
                             color='steelblue')
        axes[1].set_xlabel('Valores', fontsize=12)
        axes[1].set_ylabel('Probabilidad acumulada', fontsize=12)
        axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
        axes[1].set_ylim(-0.05, 1.05)
        axes[1].grid(alpha=0.3)
    
    # Información estadística
    info = f"""📊 Estadísticos:
    • Esperanza: μ = {esperanza:.2f}
    • Varianza: σ² = {varianza:.4f}
    • Desviación: σ = {np.sqrt(varianza):.4f}
    • Valores posibles: {n}
    • P(X=k) = 1/{n} = {1/n:.4f}
    """
    
    plt.suptitle('Distribución Uniforme Discreta', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Mostrar probabilidades específicas
    print("\n📋 Probabilidades individuales:")
    for val in x:
        print(f"  P(X = {val:3d}) = {1/n:.4f}")

# Widgets interactivos
interact(uniforme_completa,
         a=widgets.IntSlider(min=0, max=20, step=1, value=1,
                             description='Mínimo (a):'),
         b=widgets.IntSlider(min=1, max=50, step=1, value=6,
                             description='Máximo (b):'),
         mostrar_cdf=widgets.Checkbox(value=True, 
                                      description='Mostrar CDF'))
