## Distribución Weibull

### Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Weibull}(k, \lambda)$ |
| **Parámetros** | $k > 0$ = forma, $\lambda > 0$ = escala |
| **Soporte** | $[0, \infty)$ |
| **Función de densidad** | $f(x) = \frac{k}{\lambda} \left(\frac{x}{\lambda}\right)^{k-1} e^{-(x/\lambda)^k}$ |
| **Función de distribución** | $F(x) = 1 - e^{-(x/\lambda)^k}$ |
| **Esperanza** | $\mathbb{E}[X] = \lambda \Gamma(1 + 1/k)$ |
| **Varianza** | $\text{Var}(X) = \lambda^2 [\Gamma(1 + 2/k) - \Gamma(1 + 1/k)^2]$ |

###  Analogía

> "La Weibull es la **distribución de la supervivencia**. Se usa para modelar el tiempo de vida de productos, desde componentes electrónicos hasta personas."

###  Usos principales

- **Análisis de supervivencia:** Tiempo hasta fallo
- **Fiabilidad:** Vida útil de productos
- **Meteorología:** Velocidad del viento
- **Hidrología:** Precipitaciones

###  Interpretación de k (forma)

- **k < 1**: Tasa de fallo decreciente (mortalidad infantil)
- **k = 1**: Exponencial (tasa constante)
- **k > 1**: Tasa creciente (desgaste)

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import weibull_min
from scipy.special import gamma
from ipywidgets import interact, widgets



print("Kernel listo y configurado")
```

```python
# Distribución Weibull - Completa


def weibull_completa(k=1.5, lamb=1.0):
    """
    Demostración completa de la distribución Weibull
    """
    x = np.linspace(0, weibull_min.ppf(0.999, k, scale=lamb), 1000)
    pdf = weibull_min.pdf(x, k, scale=lamb)
    cdf = weibull_min.cdf(x, k, scale=lamb)
    
    esperanza = lamb * gamma(1 + 1/k)
    varianza = lamb**2 * (gamma(1 + 2/k) - gamma(1 + 1/k)**2)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf, 'b-', linewidth=2)
    axes[0].fill_between(x, pdf, alpha=0.3, color='blue')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('Densidad f(x)')
    axes[0].set_title(f'Distribución Weibull\nk = {k}, λ = {lamb}', fontsize=12)
    axes[0].axvline(x=esperanza, color='red', linestyle='--', label=f'Media = {esperanza:.3f}')
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf, alpha=0.3, color='green')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('Probabilidad acumulada F(x)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Interpretación de k
    if k < 1:
        riesgo = "tasa de fallo decreciente (mortalidad infantil)"
    elif k == 1:
        riesgo = "tasa de fallo constante (Exponencial)"
    else:
        riesgo = "tasa de fallo creciente (desgaste)"
    
    # Información
    info = f"""📊 Estadísticos:
    • Forma: k = {k} → {riesgo}
    • Escala: λ = {lamb}
    • Esperanza: μ = {esperanza:.4f}
    • Varianza: σ² = {varianza:.4f}
    • Mediana: {lamb * (np.log(2))**(1/k):.4f}
    """
    
    plt.suptitle('Distribución Weibull', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
    
    plt.tight_layout()
    plt.show()

interact(weibull_completa,
         k=widgets.FloatSlider(min=0.1, max=5, step=0.1, value=1.5,
                               description='Forma (k):'),
         lamb=widgets.FloatSlider(min=0.1, max=5, step=0.1, value=1.0,
                                  description='Escala (λ):'))
```

### 5 Aplicaciones Reales de la Distribución Weibull

1.  **Análisis de Supervivencia en Medicina:** Modelar el tiempo hasta la recaída de pacientes con cáncer después de un tratamiento y la tasa de mortalidad asociada.

2.  **Fiabilidad en Ingeniería:** Predecir la vida útil de componentes como rodamientos, ventiladores o circuitos integrados, analizando las tasas de fallo por desgaste.

3.  **Meteorología y Energía Eólica:** Modelar la velocidad del viento en una región específica para estimar el potencial de generación de energía de parques eólicos.

4.  **Hidrología:** Analizar y predecir eventos extremos como precipitaciones máximas o crecidas de ríos para el diseño de infraestructuras de control de inundaciones.

5.  **Mantenimiento Industrial:** Programar el reemplazo preventivo de piezas en maquinaria pesada analizando la distribución del tiempo de fallo de componentes mecánicos.
