## Distribución t de Student

### 📊 Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $T \sim t_{\nu}$ |
| **Parámetro** | $\nu > 0$ = grados de libertad |
| **Soporte** | $(-\infty, \infty)$ |
| **Función de densidad** | $f(t) = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\nu\pi}\,\Gamma\left(\frac{\nu}{2}\right)} \left(1+\frac{t^2}{\nu}\right)^{-\frac{\nu+1}{2}}$ |
| **Esperanza** | $\mathbb{E}[T] = 0$ (para $\nu > 1$) |
| **Varianza** | $\text{Var}(T) = \frac{\nu}{\nu-2}$ (para $\nu > 2$) |

### 🎯 Analogía

> "La t de Student es como la Normal pero con **colas más pesadas**. Es ideal cuando tienes **pocos datos** y no conoces la desviación poblacional."

### 📝 Usos principales

1. **Pruebas t:** Comparación de medias
2. **Intervalos de confianza:** Cuando σ es desconocido
3. **Regresión:** Inferencia sobre coeficientes
4. **Muestras pequeñas:** n < 30

### 🔗 Relación con Normal

A medida que $\nu \to \infty$, $t_{\nu} \to N(0,1)$

```python
# Distribución t de Student - Completa
import numpy as np
from scipy.stats import nbinom
import matplotlib.pyplot as plt # Seguramente la necesitarás para graficar
from scipy.stats import t, norm
from ipywidgets import interact, widgets

def tstudent_completa(nu=10, comparar_normal=True):
    """
    Demostración completa de la distribución t de Student
    """
    x = np.linspace(-5, 5, 1000)
    pdf_t = t.pdf(x, nu)
    cdf_t = t.cdf(x, nu)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico 1: PDF
    axes[0].plot(x, pdf_t, 'b-', linewidth=2, label=f't({nu})')
    
    if comparar_normal:
        pdf_norm = norm.pdf(x, 0, 1)
        axes[0].plot(x, pdf_norm, 'r--', linewidth=2, alpha=0.7, label='N(0,1)')
    
    axes[0].fill_between(x, pdf_t, alpha=0.3, color='blue')
    axes[0].set_xlabel('t')
    axes[0].set_ylabel('Densidad f(t)')
    axes[0].set_title(f'Distribución t de Student\nν = {nu}', fontsize=12)
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    
    # Gráfico 2: CDF
    axes[1].plot(x, cdf_t, 'g-', linewidth=2)
    axes[1].fill_between(x, cdf_t, alpha=0.3, color='green')
    axes[1].set_xlabel('t')
    axes[1].set_ylabel('Probabilidad acumulada F(t)')
    axes[1].set_title('Función de Distribución (CDF)', fontsize=12)
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(alpha=0.3)
    
    # Información
    esperanza = 0 if nu > 1 else "No definida"
    varianza = nu/(nu-2) if nu > 2 else "No definida (∞)" if nu > 2 else "No definida"
    
    info = f"""📊 Estadísticos:
    • Grados de libertad: ν = {nu}
    • Esperanza: μ = {esperanza}
    • Varianza: σ² = {varianza}
    • Curtosis: {6/(nu-4) if nu > 4 else "No definida"}
    """
    
    if comparar_normal:
        # Comparar cuantiles
        t_975 = t.ppf(0.975, nu)
        z_975 = norm.ppf(0.975, 0, 1)
        info += f"\n    • t₀.₉₇₅ = {t_975:.4f} | z₀.₉₇₅ = {z_975:.4f}"
    
    plt.suptitle('Distribución t de Student', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    print(f"\n📋 Cuantiles importantes:")
    for p in [0.9, 0.95, 0.975, 0.99]:
        print(f"  t_{p*100:.1f}% = {t.ppf(p, nu):.4f}")

interact(tstudent_completa,
         nu=widgets.IntSlider(min=1, max=100, step=1, value=10,
                              description='ν (grados):'),
         comparar_normal=widgets.Checkbox(value=True,
                                         description='Comparar con Normal'))
```
