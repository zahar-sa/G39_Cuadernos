## Esperanza Matemática (Valor Esperado)

### Resumen Rápido

| Concepto | Fórmula Discreta | Fórmula Continua |
|----------|------------------|------------------|
| **Definición** | $\mathbb{E}[X] = \sum_{i} x_i P(X=x_i)$ | $\mathbb{E}[X] = \int_{-\infty}^{\infty} x f(x) dx$ |
| **Interpretación** | Promedio ponderado | Centro de masa de la distribución |
| **Notación** | $\mu$, $\mathbb{E}[X]$, $\langle X \rangle$ | Misma notación |

###  Analogías

> **Física:** La esperanza es como el **centro de masa** de una distribución de probabilidad.
>
> **Juegos de azar:** Si juegas muchas veces, la esperanza es tu **ganancia promedio por jugada**.
>
> **Cotidiano:** Es como el **promedio** que esperas obtener a largo plazo.

###  Ejemplos intuitivos

**Ejemplo 1 - Dado justo:**
- Valores: 1, 2, 3, 4, 5, 6 (cada uno con probabilidad 1/6)
- $\mathbb{E}[X] = (1+2+3+4+5+6)/6 = 3.5$

**Ejemplo 2 - Lotería:**
- Premio: $1,000,000 con probabilidad 1/1,000,000
- Nada con probabilidad 999,999/1,000,000
- $\mathbb{E}[X] = 1,000,000 \times 0.000001 + 0 \times 0.999999 = \$1$

**Interpretación:** ¡En promedio, solo ganas $1 por boleto!

###  Propiedades fundamentales

1. **Linealidad:** $\mathbb{E}[aX + bY] = a\mathbb{E}[X] + b\mathbb{E}[Y]$
2. **Constante:** $\mathbb{E}[c] = c$
3. **Monotonicidad:** Si $X \leq Y$, entonces $\mathbb{E}[X] \leq \mathbb{E}[Y]$
4. **Independencia:** Si $X \perp Y$, $\mathbb{E}[XY] = \mathbb{E}[X]\mathbb{E}[Y]$

###  Ley de los Grandes Números (LGN)

> El promedio de una muestra se acerca a la esperanza cuando el tamaño de muestra crece:

$$\bar{X}_n = \frac{1}{n}\sum_{i=1}^n X_i \xrightarrow{p} \mathbb{E}[X]$$

**Interpretación:** "La frecuencia se convierte en probabilidad cuando hay muchos datos."

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm, poisson
from ipywidgets import interact, widgets



print("Kernel listo y configurado")
```

```python
# Esperanza Matemática - Demostración Interactiva


def demostracion_esperanza(distribucion="Binomial", n=30, p=0.5, lamb=3, mu=0, sigma=1, n_muestras=1000):
    """
    Demostración interactiva de la esperanza matemática
    """
    np.random.seed(42)
    
    # Seleccionar distribución
    if distribucion == "Binomial":
        X = binom(n, p)
        esperanza_teorica = n * p
        muestras = binom.rvs(n, p, size=n_muestras)
        titulo = f'Binomial(n={n}, p={p})'
        x_vals = np.arange(0, n+1)
        pmf = binom.pmf(x_vals, n, p)
        
    elif distribucion == "Poisson":
        X = poisson(lamb)
        esperanza_teorica = lamb
        muestras = poisson.rvs(lamb, size=n_muestras)
        titulo = f'Poisson(λ={lamb})'
        x_vals = np.arange(0, 20)
        pmf = poisson.pmf(x_vals, lamb)
        
    else:  # Normal
        esperanza_teorica = mu
        muestras = norm.rvs(mu, sigma, size=n_muestras)
        titulo = f'Normal(μ={mu}, σ={sigma})'
        x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)
        pmf = norm.pdf(x_vals, mu, sigma)
    
    # Crear figura
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # Gráfico 1: Distribución teórica
    if distribucion in ["Binomial", "Poisson"]:
        axes[0,0].bar(x_vals, pmf, width=0.6, color='skyblue', edgecolor='black')
        axes[0,0].axvline(x=esperanza_teorica, color='red', linewidth=2, 
                         label=f'μ = {esperanza_teorica:.3f}')
    else:
        axes[0,0].plot(x_vals, pmf, 'b-', linewidth=2)
        axes[0,0].fill_between(x_vals, pmf, alpha=0.3, color='blue')
        axes[0,0].axvline(x=esperanza_teorica, color='red', linewidth=2,
                         label=f'μ = {esperanza_teorica:.3f}')
    
    axes[0,0].set_title(f'Distribución Teórica\n{titulo}', fontsize=12)
    axes[0,0].set_xlabel('x')
    axes[0,0].set_ylabel('Probabilidad/Densidad')
    axes[0,0].legend()
    axes[0,0].grid(alpha=0.3)
    
    # Gráfico 2: Convergencia de la media muestral
    medias_acum = np.cumsum(muestras) / np.arange(1, n_muestras + 1)
    axes[0,1].plot(range(1, n_muestras + 1), medias_acum, 'b-', linewidth=0.5, alpha=0.7)
    axes[0,1].axhline(y=esperanza_teorica, color='red', linestyle='--', 
                     linewidth=2, label=f'μ teórica = {esperanza_teorica:.3f}')
    axes[0,1].set_xlabel('Número de muestras')
    axes[0,1].set_ylabel('Media acumulada')
    axes[0,1].set_title('Ley de los Grandes Números\nConvergencia de la media', fontsize=12)
    axes[0,1].legend()
    axes[0,1].grid(alpha=0.3)
    
    # Gráfico 3: Histograma de las muestras
    axes[1,0].hist(muestras, bins=30, density=True, alpha=0.7, color='lightgreen', edgecolor='black')
    axes[1,0].axvline(x=esperanza_teorica, color='red', linewidth=2, 
                     label=f'μ teórica = {esperanza_teorica:.3f}')
    axes[1,0].axvline(x=np.mean(muestras), color='blue', linestyle='--', linewidth=2,
                     label=f'μ muestral = {np.mean(muestras):.3f}')
    axes[1,0].set_xlabel('x')
    axes[1,0].set_ylabel('Densidad')
    axes[1,0].set_title(f'Histograma de {n_muestras} muestras', fontsize=12)
    axes[1,0].legend()
    axes[1,0].grid(alpha=0.3)
    
    # Gráfico 4: Error de estimación
    error = np.abs(medias_acum - esperanza_teorica)
    axes[1,1].plot(range(1, n_muestras + 1), error, 'r-', linewidth=0.5, alpha=0.7)
    axes[1,1].set_xlabel('Número de muestras')
    axes[1,1].set_ylabel('Error absoluto')
    axes[1,1].set_title('Error de estimación\n|μ_muestral - μ_teórica|', fontsize=12)
    axes[1,1].set_yscale('log')
    axes[1,1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Resultados:
    • Esperanza teórica: μ = {esperanza_teorica:.4f}
    • Esperanza muestral: x̄ = {np.mean(muestras):.4f}
    • Error absoluto: {np.abs(np.mean(muestras) - esperanza_teorica):.6f}
    • Desviación estándar muestral: s = {np.std(muestras):.4f}
    """
    
    plt.suptitle('Demostración de la Esperanza Matemática', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Interpretación
    print("\n📌 Interpretación:")
    print(f"  El valor esperado de {n_muestras} observaciones es {np.mean(muestras):.4f},")
    print(f"  mientras que el valor teórico es {esperanza_teorica:.4f}.")
    print(f"  La diferencia se reduce a medida que aumentan las muestras.")

interact(demostracion_esperanza,
         distribucion=widgets.Dropdown(
             options=['Binomial', 'Poisson', 'Normal'],
             value='Binomial',
             description='Distribución:'),
         n=widgets.IntSlider(min=1, max=100, step=1, value=30,
                            description='n (Binomial):'),
         p=widgets.FloatSlider(min=0, max=1, step=0.01, value=0.5,
                              description='p (Binomial):'),
         lamb=widgets.FloatSlider(min=0.5, max=20, step=0.5, value=5,
                                 description='λ (Poisson):'),
         mu=widgets.FloatSlider(min=-5, max=5, step=0.5, value=0,
                               description='μ (Normal):'),
         sigma=widgets.FloatSlider(min=0.1, max=3, step=0.1, value=1,
                                  description='σ (Normal):'),
         n_muestras=widgets.IntSlider(min=100, max=10000, step=100, value=1000,
                                     description='N° muestras:'))
```
