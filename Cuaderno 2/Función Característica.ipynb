## Función Característica

###  Definición


$$\phi_X(t) = \mathbb{E}[e^{itX}] = \mathbb{E}[\cos(tX)] + i\mathbb{E}[\sin(tX)]$$

donde $i = \sqrt{-1}$ es la unidad imaginaria.

###  Ventajas sobre la MGF

| Propiedad | MGF $M_X(t)$ | Función Característica $\phi_X(t)$ |
|-----------|--------------|-------------------------------------|
| **Existencia** | No siempre existe | **Siempre existe** |
| **Dominio** | Puede ser restringido | $t \in \mathbb{R}$ |
| **Complejidad** | Real | Compleja |
| **Unicidad** | Sí | Sí |

###  Ejemplos de Funciones Características

| Distribución | $\phi_X(t)$ |
|--------------|--------------|
| **Bernoulli(p)** | $1-p + pe^{it}$ |
| **Binomial(n, p)** | $(1-p + pe^{it})^n$ |
| **Poisson(λ)** | $e^{\lambda(e^{it} - 1)}$ |
| **Normal(μ, σ²)** | $e^{i\mu t - \frac{1}{2}\sigma^2 t^2}$ |
| **Exponencial(λ)** | $\frac{\lambda}{\lambda - it}$ |

###  Propiedades clave

1. **Unicidad:** Dos distribuciones son iguales si y solo si tienen la misma función característica.
2. **Continuidad:** Es uniformemente continua en $\mathbb{R}$
3. **Inversión:** Permite recuperar la distribución:
   $$f(x) = \frac{1}{2\pi} \int_{-\infty}^{\infty} e^{-itx} \phi_X(t) dt$$

4. **Teorema de Lévy:** La convergencia en distribución es equivalente a la convergencia puntual de las funciones características.

###  ¿Por qué es fundamental?

- **Siempre existe** (a diferencia de la MGF)
- **Prueba del TCL** (Teorema Central del Límite)
- **Estabilidad** de distribuciones
- **Procesos estocásticos** y ecuaciones diferenciales

## Distribución t de Student

###  Resumen Rápido

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $T \sim t_{\nu}$ |
| **Parámetro** | $\nu > 0$ = grados de libertad |
| **Soporte** | $(-\infty, \infty)$ |
| **Función de densidad** | $f(t) = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\nu\pi}\,\Gamma\left(\frac{\nu}{2}\right)} \left(1+\frac{t^2}{\nu}\right)^{-\frac{\nu+1}{2}}$ |
| **Esperanza** | $\mathbb{E}[T] = 0$ (para $\nu > 1$) |
| **Varianza** | $\text{Var}(T) = \frac{\nu}{\nu-2}$ (para $\nu > 2$) |

###  Analogía

> "La t de Student es como la Normal pero con **colas más pesadas**. Es ideal cuando tienes **pocos datos** y no conoces la desviación poblacional."

###  Usos principales

1. **Pruebas t:** Comparación de medias
2. **Intervalos de confianza:** Cuando σ es desconocido
3. **Regresión:** Inferencia sobre coeficientes
4. **Muestras pequeñas:** n < 30

###  Relación con Normal

A medida que $\nu \to \infty$, $t_{\nu} \to N(0,1)$

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, cauchy
from ipywidgets import interact, widgets



print("Kernel listo y configurado")
```

```python
# Función Característica - Demostración


def phi_normal(t, mu=0, sigma=1):
    """Función característica de la Normal"""
    return np.exp(1j*mu*t - 0.5*sigma**2*t**2)

def phi_exponencial(t, lamb=1):
    """Función característica de la Exponencial"""
    return lamb/(lamb - 1j*t)

def phi_laplace(t, mu=0, b=1):
    """Función característica de Laplace"""
    return np.exp(1j*mu*t) / (1 + b**2*t**2)

def phi_cauchy(t, x0=0, gamma=1):
    """Función característica de Cauchy"""
    return np.exp(1j*x0*t - gamma*np.abs(t))

def demostracion_funcion_caracteristica(distribucion="Normal", mu=0, sigma=1, lamb=1, t_max=5):
    """
    Demostración de la Función Característica
    """
    t_vals = np.linspace(-t_max, t_max, 500)
    
    # Seleccionar función característica
    if distribucion == "Normal":
        phi = phi_normal(t_vals, mu, sigma)
        titulo = f'N({mu}, {sigma}²)'
        pdf = lambda x: norm.pdf(x, mu, sigma)
        x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
        
    elif distribucion == "Exponencial":
        phi = phi_exponencial(t_vals, lamb)
        titulo = f'Exp(λ={lamb})'
        pdf = lambda x: lamb * np.exp(-lamb*x) if x >= 0 else 0
        x_vals = np.linspace(0, 5/lamb, 500)
        
    elif distribucion == "Laplace":
        phi = phi_laplace(t_vals, mu, lamb)
        titulo = f'Laplace(μ={mu}, b={lamb})'
        pdf = lambda x: (1/(2*lamb)) * np.exp(-np.abs(x-mu)/lamb)
        x_vals = np.linspace(mu - 5*lamb, mu + 5*lamb, 500)
        
    else:  # Cauchy
        phi = phi_cauchy(t_vals, mu, sigma)
        titulo = f'Cauchy(x₀={mu}, γ={sigma})'
        pdf = lambda x: 1/(np.pi*sigma*(1 + ((x-mu)/sigma)**2))
        x_vals = np.linspace(mu - 10*sigma, mu + 10*sigma, 500)
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # Gráfico 1: Parte Real de φ(t)
    axes[0,0].plot(t_vals, np.real(phi), 'b-', linewidth=2, label='Re[φ(t)]')
    axes[0,0].set_xlabel('t')
    axes[0,0].set_ylabel('Parte Real')
    axes[0,0].set_title(f'Función Característica - Parte Real\n{distribucion}: {titulo}', fontsize=12)
    axes[0,0].grid(alpha=0.3)
    axes[0,0].legend()
    
    # Gráfico 2: Parte Imaginaria de φ(t)
    axes[0,1].plot(t_vals, np.imag(phi), 'r-', linewidth=2, label='Im[φ(t)]')
    axes[0,1].set_xlabel('t')
    axes[0,1].set_ylabel('Parte Imaginaria')
    axes[0,1].set_title('Función Característica - Parte Imaginaria', fontsize=12)
    axes[0,1].grid(alpha=0.3)
    axes[0,1].legend()
    
    # Gráfico 3: Módulo de φ(t)
    modulo = np.abs(phi)
    axes[1,0].plot(t_vals, modulo, 'g-', linewidth=2)
    axes[1,0].set_xlabel('t')
    axes[1,0].set_ylabel('|φ(t)|')
    axes[1,0].set_title('Módulo de la Función Característica', fontsize=12)
    axes[1,0].set_ylim(-0.05, 1.05)
    axes[1,0].grid(alpha=0.3)
    
    # Gráfico 4: PDF correspondiente
    pdf_vals = np.array([pdf(x) for x in x_vals])
    axes[1,1].plot(x_vals, pdf_vals, 'purple', linewidth=2)
    axes[1,1].fill_between(x_vals, pdf_vals, alpha=0.3, color='purple')
    axes[1,1].set_xlabel('x')
    axes[1,1].set_ylabel('f(x)')
    axes[1,1].set_title('Función de Densidad de Probabilidad (PDF)', fontsize=12)
    axes[1,1].grid(alpha=0.3)
    
    # Información
    info = f"""📌 Propiedades de la Función Característica:
    • φ(0) = {np.abs(phi[250]):.4f} (debe ser 1)
    • |φ(t)| ≤ 1 para todo t
    • Existe SIEMPRE (a diferencia de la MGF)
    • Permite recuperar la distribución por inversión de Fourier
    • Fundamental para el Teorema Central del Límite
    """
    
    plt.suptitle('Función Característica', fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    print("\n📊 Comparación MGF vs Función Característica:")
    print("  • MGF: Puede no existir (ej: Cauchy, Laplace)")
    print("  • Función Característica: SIEMPRE existe")
    print("  • Ambas identifican la distribución unívocamente")

interact(demostracion_funcion_caracteristica,
         distribucion=widgets.Dropdown(
             options=['Normal', 'Exponencial', 'Laplace', 'Cauchy'],
             value='Normal',
             description='Distribución:'),
         mu=widgets.FloatSlider(min=-3, max=3, step=0.5, value=0,
                               description='μ / x₀:'),
         sigma=widgets.FloatSlider(min=0.1, max=2, step=0.1, value=1,
                                  description='σ / γ / b:'),
         lamb=widgets.FloatSlider(min=0.1, max=3, step=0.1, value=1,
                                 description='λ (Exponencial):'),
         t_max=widgets.FloatSlider(min=2, max=10, step=0.5, value=5,
                                  description='Rango de t:'))
```

### 5 Aplicaciones Reales de la Función Característica

1.  **Suma de Variables Independientes:** Permite calcular la distribución de la suma de variables aleatorias independientes (por ejemplo, el ingreso total de una cartera de inversiones) multiplicando sus funciones características, simplificando un problema que sería complejo con convoluciones.

2.  **Demostración del Teorema Central del Límite (TCL):** Es la herramienta matemática fundamental para demostrar el TCL, que justifica por qué la media de muchas variables (como errores de medición en un proceso industrial) sigue una distribución normal.

3.  **Valoración de Opciones Financieras:** Se utiliza en modelos matemáticos avanzados, como ciertos modelos de difusión con saltos (ej. modelos de Kou), para describir movimientos de precios de activos que no siguen un comportamiento puramente normal.

4.  **Procesamiento de Señales:** En telecomunicaciones, la función característica (relacionada con la transformada de Fourier de la densidad de probabilidad) se usa para analizar el ruido y la interferencia en señales digitales.

5.  **Identificación de Distribuciones:** Sirve para identificar de forma única la distribución de un fenómeno aleatorio complejo en hidrología (ej. caudales de ríos) cuando otras herramientas (como la función generadora de momentos) no existen.
