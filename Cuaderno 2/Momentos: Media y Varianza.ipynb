## Momentos de una Distribución

###  Definición de Momentos

| Tipo | Fórmula | Interpretación |
|------|---------|----------------|
| **Momento de orden 1** (Media) | $\mu = \mathbb{E}[X]$ | Centro de la distribución |
| **Momento centrado de orden 2** (Varianza) | $\sigma^2 = \mathbb{E}[(X-\mu)^2]$ | Dispersión alrededor de la media |
| **Momento de orden k** | $\mu_k = \mathbb{E}[X^k]$ | Momentos no centrados |
| **Momento centrado de orden k** | $m_k = \mathbb{E}[(X-\mu)^k]$ | Momentos alrededor de la media |

###  Analogías

> **Media (μ):** Es como el **centro de gravedad** de la distribución.
>
> **Varianza (σ²):** Mide **cuán esparcidos** están los datos. Es como el **momento de inercia** en física.
>
> **Desviación estándar (σ):** Es la raíz cuadrada de la varianza. Está en las **mismas unidades** que los datos.

````python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon, uniform, skew, kurtosis
from ipywidgets import interact, widgets  # ← Línea faltante
import warnings
warnings.filterwarnings('ignore')



print("Kernel listo y configurado")
````

````python

### Nivel 2: Código Interactivo

# Momentos: Media y Varianza - Demostración Interactiva


def demostracion_momentos(distribucion="Normal", param1=0, param2=1, n_muestras=1000):
    """
    Demostración interactiva de momentos (media, varianza, asimetría, curtosis)
    """
    np.random.seed(42)
    
    # Seleccionar distribución
    if distribucion == "Normal":
        mu, sigma = param1, param2
        muestras = norm.rvs(mu, sigma, size=n_muestras)
        x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)
        pdf = norm.pdf(x_vals, mu, sigma)
        titulo = f'N({mu}, {sigma}²)'
        teorica = {
            'media': mu,
            'varianza': sigma**2,
            'asimetria': 0,
            'curtosis': 0
        }
        
    elif distribucion == "Exponencial":
        lamb = param1
        muestras = np.random.exponential(1/lamb, size=n_muestras)
        x_vals = np.linspace(0, 5/lamb, 100)
        pdf = lamb * np.exp(-lamb * x_vals)
        titulo = f'Exp(λ={lamb})'
        teorica = {
            'media': 1/lamb,
            'varianza': 1/lamb**2,
            'asimetria': 2,
            'curtosis': 6
        }
        
    else:  # Uniforme
        a, b = param1, param2
        muestras = np.random.uniform(a, b, size=n_muestras)
        x_vals = np.linspace(a - 0.1*(b-a), b + 0.1*(b-a), 100)
        pdf = np.ones_like(x_vals) / (b-a)
        pdf[(x_vals < a) | (x_vals > b)] = 0
        titulo = f'U({a}, {b})'
        teorica = {
            'media': (a+b)/2,
            'varianza': (b-a)**2/12,
            'asimetria': 0,
            'curtosis': -1.2
        }
    
    # Calcular momentos muestrales
    media_muestral = np.mean(muestras)
    varianza_muestral = np.var(muestras)
    asimetria_muestral = skew(muestras)
    curtosis_muestral = kurtosis(muestras)
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # Gráfico 1: Distribución con media y desviación
    axes[0,0].plot(x_vals, pdf, 'b-', linewidth=2, label='PDF teórica')
    axes[0,0].fill_between(x_vals, pdf, alpha=0.3, color='blue')
    axes[0,0].axvline(x=teorica['media'], color='red', linewidth=2, 
                     label=f'μ = {teorica["media"]:.3f}')
    axes[0,0].axvline(x=teorica['media'] - np.sqrt(teorica['varianza']), 
                     color='orange', linestyle='--', alpha=0.7,
                     label=f'μ ± σ')
    axes[0,0].axvline(x=teorica['media'] + np.sqrt(teorica['varianza']), 
                     color='orange', linestyle='--', alpha=0.7)
    axes[0,0].set_title(f'Distribución Teórica\n{titulo}', fontsize=12)
    axes[0,0].set_xlabel('x')
    axes[0,0].set_ylabel('Densidad')
    axes[0,0].legend()
    axes[0,0].grid(alpha=0.3)
    
    # Gráfico 2: Histograma de las muestras
    axes[0,1].hist(muestras, bins=40, density=True, alpha=0.7, 
                   color='lightgreen', edgecolor='black', label='Muestras')
    axes[0,1].axvline(x=media_muestral, color='red', linewidth=2,
                     label=f'x̄ = {media_muestral:.3f}')
    axes[0,1].axvline(x=media_muestral - np.std(muestras), color='orange', 
                     linestyle='--', alpha=0.7)
    axes[0,1].axvline(x=media_muestral + np.std(muestras), color='orange', 
                     linestyle='--', alpha=0.7)
    axes[0,1].set_title(f'Histograma (n={n_muestras})', fontsize=12)
    axes[0,1].set_xlabel('x')
    axes[0,1].set_ylabel('Densidad')
    axes[0,1].legend()
    axes[0,1].grid(alpha=0.3)
    
    # Gráfico 3: Comparación de momentos
    categorias = ['Media', 'Varianza', 'Asimetría', 'Curtosis']
    valores_teoricos = [teorica['media'], teorica['varianza'], 
                        teorica['asimetria'], teorica['curtosis']]
    valores_muestrales = [media_muestral, varianza_muestral, 
                          asimetria_muestral, curtosis_muestral]
    
    x_pos = np.arange(len(categorias))
    width = 0.35
    
    axes[1,0].bar(x_pos - width/2, valores_teoricos, width, 
                 label='Teórico', color='skyblue', edgecolor='black')
    axes[1,0].bar(x_pos + width/2, valores_muestrales, width,
                 label='Muestral', color='lightcoral', edgecolor='black')
    axes[1,0].set_xticks(x_pos)
    axes[1,0].set_xticklabels(categorias)
    axes[1,0].set_title('Comparación de Momentos', fontsize=12)
    axes[1,0].legend()
    axes[1,0].grid(axis='y', alpha=0.3)
    
    # Gráfico 4: Boxplot para visualizar dispersión
    bp = axes[1,1].boxplot(muestras, vert=True, patch_artist=True)
    bp['boxes'][0].set_facecolor('lightblue')
    axes[1,1].set_title('Diagrama de Caja (Boxplot)\nVisualización de dispersión', fontsize=12)
    axes[1,1].set_ylabel('Valores')
    axes[1,1].grid(alpha=0.3)
    
    # Información
    info = f"""📊 Momentos:
    • Media: μ = {teorica['media']:.4f} | x̄ = {media_muestral:.4f}
    • Varianza: σ² = {teorica['varianza']:.4f} | s² = {varianza_muestral:.4f}
    • Desv. Estándar: σ = {np.sqrt(teorica['varianza']):.4f} | s = {np.std(muestras):.4f}
    • Asimetría: γ₁ = {teorica['asimetria']:.4f} | g₁ = {asimetria_muestral:.4f}
    • Curtosis: γ₂ = {teorica['curtosis']:.4f} | g₂ = {curtosis_muestral:.4f}
    """
    
    plt.suptitle('Momentos de una Distribución: Media, Varianza, Asimetría, Curtosis', 
                fontsize=14, y=1.02)
    fig.text(0.5, -0.02, info, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
    
    # Interpretación de asimetría y curtosis
    print("\n📌 Interpretación:")
    if abs(asimetria_muestral) < 0.5:
        print(f"  • Asimetría ({asimetria_muestral:.3f}): Distribución aproximadamente simétrica")
    elif asimetria_muestral > 0:
        print(f"  • Asimetría ({asimetria_muestral:.3f}): Sesgo positivo (cola derecha más larga)")
    else:
        print(f"  • Asimetría ({asimetria_muestral:.3f}): Sesgo negativo (cola izquierda más larga)")
    
    if abs(curtosis_muestral) < 0.5:
        print(f"  • Curtosis ({curtosis_muestral:.3f}): Colas similares a la Normal")
    elif curtosis_muestral > 0:
        print(f"  • Curtosis ({curtosis_muestral:.3f}): Colas pesadas (leptocúrtica)")
    else:
        print(f"  • Curtosis ({curtosis_muestral:.3f}): Colas ligeras (platicúrtica)")

interact(demostracion_momentos,
         distribucion=widgets.Dropdown(
             options=['Normal', 'Exponencial', 'Uniforme'],
             value='Normal',
             description='Distribución:'),
         param1=widgets.FloatSlider(min=-3, max=3, step=0.1, value=0,
                                   description='Parámetro 1:'),
         param2=widgets.FloatSlider(min=0.1, max=3, step=0.1, value=1,
                                   description='Parámetro 2:'),
         n_muestras=widgets.IntSlider(min=100, max=5000, step=100, value=1000,
                                     description='N° muestras:'))
````
