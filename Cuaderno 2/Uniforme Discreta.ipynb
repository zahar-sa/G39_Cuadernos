 📊 **Resumen Rápido: Uniforme Discreta**

| Característica | Descripción |
|----------------|-------------|
| **Notación** | $X \sim \text{Uniforme}(x_1, x_2, \dots, x_n)$ o $X \sim \text{Unif}\{a, a+1, \dots, b\}$ |
| **Parámetros** | $a$ (mínimo), $b$ (máximo), $n = b-a+1$ (número de valores posibles) |
| **Espacio muestral** | $\{a, a+1, a+2, \dots, b\}$ |
| **Función de masa (pmf)** | $P(X=k) = \frac{1}{n}$ para $k \in \{a, a+1, \dots, b\}$ |
| **Esperanza** | $\mathbb{E}[X] = \frac{a+b}{2}$ |
| **Varianza** | $\text{Var}(X) = \frac{(b-a+1)^2 - 1}{12}$ |
| **Usos típicos** | Lanzamiento de un dado justo, rifas, selección aleatoria simple, números aleatorios |

### 🎲 Analogía

> "La distribución uniforme discreta es como tener una ruleta con $n$ números igualmente espaciados, donde cada número tiene **exactamente la misma probabilidad** de salir. Es el modelo de la 'suerte perfecta' donde no hay favoritismo por ningún valor."

### 🔍 ¿Cuándo se usa?

- **Juegos de azar:** Dado justo (6 caras), ruleta (37 o 38 números)
- **Muestreo aleatorio:** Seleccionar un elemento al azar de una lista
- **Rifas y sorteos:** Cada boleto tiene la misma probabilidad de ganar
- **Simulaciones:** Generar números enteros aleatorios en un rango

$$Código

```python
# Distribución Uniforme Discreta - Versión Interactiva
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint
from ipywidgets import interact, widgets

def graficar_uniforme_interactiva(min_val=0, max_val=20):
    """
    Grafica la función de masa de probabilidad (pmf) 
    de una distribución uniforme discreta
    """
    # Generar valores posibles
    x = np.arange(min_val, max_val + 1)
    n_valores = len(x)
    
    # Calcular pmf (cada valor tiene la misma probabilidad)
    pmf_unif = np.full_like(x, 1/n_valores, dtype=float)
    
    # Crear gráfica
    plt.figure(figsize=(10, 5))
    
    # Barras de la distribución
    plt.bar(x, pmf_unif, width=0.6, color='coral', 
            edgecolor='black', alpha=0.8, label='Probabilidad')
    
    # Línea horizontal para mostrar que es constante
    plt.axhline(y=1/n_valores, color='red', linestyle='--', 
                alpha=0.5, label=f'Constante = {1/n_valores:.4f}')
    
    # Decoraciones
    plt.xlabel('Valores posibles', fontsize=12)
    plt.ylabel('Probabilidad', fontsize=12)
    plt.title(f'Distribución Uniforme Discreta\n' +
              f'Valores: [{min_val}, {max_val}] - ' +
              f'{n_valores} valores igualmente probables', 
              fontsize=14)
    plt.xticks(x, rotation=45)
    plt.ylim(0, max(pmf_unif) * 1.2)
    plt.grid(axis='y', alpha=0.3, linestyle='--')
    plt.legend()
    
    # Mostrar información adicional
    esperanza = (min_val + max_val) / 2
    varianza = ((max_val - min_val + 1)**2 - 1) / 12
    texto_info = f"Esperanza: {esperanza:.2f} | Varianza: {varianza:.2f}"
    plt.figtext(0.5, -0.05, texto_info, ha='center', fontsize=10)
    
    plt.tight_layout()
    plt.show()

# Ejecutar el widget interactivo
interact(graficar_uniforme_interactiva,
         min_val=widgets.IntSlider(min=0, max=50, step=1, value=0,
                                   description='Mínimo:'),
         max_val=widgets.IntSlider(min=1, max=100, step=1, value=20,
                                   description='Máximo:'))
```
