##  Fenómenos Deterministas vs Aleatorios

### ¿Cuál es la diferencia?

| Característica | Fenómeno Determinista | Fenómeno Aleatorio |
|----------------|----------------------|-------------------|
| **Definición** | El resultado se puede predecir con certeza | El resultado no se puede predecir con certeza |
| **Ejemplo** | Lanzar una piedra al aire (cae) | Lanzar un dado (no sé qué número saldrá) |
| **Variables** | $F = m \cdot a$ (Ley de Newton) | $X$ = Número al lanzar un dado |
| **Repetibilidad** | Mismo inicio → mismo final | Mismo inicio → resultados diferentes |
| **Modelo** | Ecuaciones exactas | Distribuciones de probabilidad |

###  Analogía

> **Determinista:** Es como seguir una receta de cocina al pie de la letra → siempre obtienes el mismo platillo.
>
> **Aleatorio:** Es como jugar a la lotería → compras el boleto pero no sabes si ganarás.

###  Ejemplos cotidianos

**Fenómenos Deterministas:**
- El Sol sale por el Este todas las mañanas
- Si calientas agua a 100°C al nivel del mar, hierve
- Un objeto lanzado hacia arriba siempre cae

**Fenómenos Aleatorios:**
- El resultado de un partido de fútbol
- El tiempo que hará mañana
- El número de clientes que entrarán a una tienda hoy

###  Relación con las variables aleatorias

Las **variables aleatorias** son funciones que **cuantifican** fenómenos aleatorios:



###  Variable Aleatoria (Definición formal)

> Una **variable aleatoria** es una función que asigna un número real a cada resultado posible de un experimento aleatorio.

**Notación:** $X: \Omega \rightarrow \mathbb{R}$

- $\Omega$: Espacio muestral (todos los resultados posibles)
- $X(\omega)$: Valor numérico asociado al resultado $\omega$

**Ejemplo:** Lanzar dos monedas
- $\Omega = \{CC, CS, SC, SS\}$ (C=Cara, S=Sello)
- $X$ = Número de caras → $X(CC)=2, X(CS)=1, X(SC)=1, X(SS)=0$

```python
# prueba de inicio para verificar el kernel y las bibliotecas necesarias

%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets


print("Kernel listo y configurado")
```

```python
# Demostración interactiva: Determinista vs Aleatorio


def comparacion_determinista_aleatorio(n_lanzamientos=100):
    """
    Compara un fenómeno determinista (caída libre) 
    vs uno aleatorio (lanzamiento de moneda)
    """
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # === FENÓMENO DETERMINISTA: Caída libre ===
    t = np.linspace(0, 2, 100)
    g = 9.81  # gravedad (m/s²)
    h0 = 10   # altura inicial (m)
    h = h0 - 0.5 * g * t**2
    h = np.maximum(h, 0)  # no puede ser negativo
    
    ax1.plot(t, h, 'b-', linewidth=2)
    ax1.fill_between(t, h, alpha=0.3, color='blue')
    ax1.set_xlabel('Tiempo (segundos)')
    ax1.set_ylabel('Altura (metros)')
    ax1.set_title('Fenómeno Determinista: Caída Libre\nMismo inicio → Mismo final', fontsize=12)
    ax1.grid(True, alpha=0.3)
    ax1.axhline(y=0, color='red', linestyle='--', alpha=0.5)
    
    # === FENÓMENO ALEATORIO: Lanzamiento de moneda ===
    resultados = np.random.choice(['Cara', 'Sello'], size=n_lanzamientos)
    caras = np.cumsum(resultados == 'Cara')
    proporcion_caras = caras / np.arange(1, n_lanzamientos + 1)
    
    ax2.plot(range(1, n_lanzamientos + 1), proporcion_caras, 
             'g-', alpha=0.7, label='Proporción observada')
    ax2.axhline(y=0.5, color='red', linestyle='--', 
                label='Probabilidad teórica (0.5)')
    ax2.set_xlabel('Número de lanzamientos')
    ax2.set_ylabel('Proporción de caras')
    ax2.set_title('Fenómeno Aleatorio: Moneda\nResultado impredecible', fontsize=12)
    ax2.set_ylim(0, 1)
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    
    # Información adicional
    ultima_proporcion = proporcion_caras[-1]
    info_text = f"""Resultados después de {n_lanzamientos} lanzamientos:
    • Caras: {caras[-1]}
    • Sellos: {n_lanzamientos - caras[-1]}
    • Proporción de caras: {ultima_proporcion:.3f}
    • Diferencia con 0.5: {abs(ultima_proporcion - 0.5):.3f}
    """
    
    plt.suptitle('Determinista vs Aleatorio', fontsize=16, y=1.02)
    fig.text(0.5, -0.02, info_text, ha='center', fontsize=10,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()

interact(comparacion_determinista_aleatorio,
         n_lanzamientos=widgets.IntSlider(min=10, max=1000, step=10, value=100,
                                          description='Lanzamientos:'))
```
